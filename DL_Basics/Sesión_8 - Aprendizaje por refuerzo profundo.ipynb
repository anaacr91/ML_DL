## 8.1.- Aprendizaje por refuerzo mediante redes neuronales

### 8.1.1.- Deep Q-Networks

En las Deep Q-networks se reemplazan las tablas de Q(s,a) por redes neuronales profundas. La estructura habitual es tomar un espacio de estados como entrada de la red neuronal y el vector de acciones como salida, de modo que, para cada vectorización de los estados del entorno "s", dispondremos de una estimación/aproximación de "Q" para cada acción del agente, que quedarán activadas mediante funciones softmax.

La Deep Q-Network original se utilizó para derrotar en 2015 a múltimples juegos Atari de forma descentralizada únicamente partiendo de información visual (redes convolucionales para tratar los pixels de la imagen como estado del entorno), pero con el tiempo se ha ido mejorando la arquitectura de la red. Estos son algunos de los avances:

* **Double DQN**: El problema con el algoritmo DQN es que sobreestima las recompensas reales, de modo que los valores-Q que aprende piensan que van a obtener una recompensa mayor de lo que en realidad obtendrá. Para solucionarlo, los autores del algoritmo Double DQN proponen un sencillo truco: separar la selección y evaluación de la acción en dos pasos diferentes. 
* **Dueling DQN**: Este algoritmo divide los valores-Q en dos partes distintas, la función de valor (value function) V(s) y la función de ventaja (advantage function) A(s, a). La función de valor V(s) nos dice cuánta recompensa obtendremos desde el estado s. La función de ventaja A(s, a) nos dice cuánto mejor es una acción respecto a las demás. 
* **Rainbow DQN**: combina varias mejoras, como Double Q-Learning para abordar el sesgo de sobreestimación, repetición de experiencia priorizada para priorizar transiciones importantes, uso de dueling networks, aprendizaje de varios pasos, aprendizaje por refuerzo distribucional en lugar del rendimiento esperado y capas lineales ruidosas para la exploración.

### 8.1.2.- Deep Policy Gradients

Los modelos básicos (apodados "vanilla") de Deep Policy Gradient basados en aprendizaje profundo consisten en utilizar una red neuronal para representar la policy, es decir, el mecanismo para seleccionar la siguiente acción a realizar. Una importante ventaja que poseen los modelos de policy gradients sobre los DQN es que son válidos tanto para la selección de acciones concretas como para la selección de acciones representadas por vectores continuos como, por ejemplo, un valor de inclinación o de velocidad continuo y preciso.

Entre las técnicas modernas para implementar este tipo de modelos mediante aprendizzaje profundo destacan las siguientes:
* **Trust Region Policy Optimization (TRPO)**: no es tan diferente de los primeros algoritmos de policy gradients, pero aporta tres mejoras principales: el uso del método del gradiente conjugado, la búsqueda de línea y el control de mejora. Cada una de estas características aborda un problema en el algoritmo original.
* **Proximal Policy Optimization (PPO)**: es una familia de algoritmos de aprendizaje por refuerzo sin modelo desarrollados en OpenAI en 2017. Los algoritmos PPO son métodos de gradiente de políticas, lo que significa que buscan en el espacio de políticas en lugar de asignar valores a pares de estado-acción.
* **Deep Deterministic Policy Gradient (DDPG)**: es un algoritmo que aprende simultáneamente una función Q y una política. Utiliza datos fuera de la política y la ecuación de Bellman para aprender la función Q y usa la función Q para aprender la política. una subrutina dolorosamente costosa.
* **Twin Delayed Deep Deterministic Policy Gradient (TD3)**: es un método de aprendizaje de refuerzo off-policy, en línea y sin modelo. Un agente TD3 es un agente de aprendizaje por refuerzo actor-crítico que busca una política óptima que maximice la recompensa acumulada esperada a largo plazo.
* **Importance Weighted Actor-Learner Architecture (IMPALA)**: no solo usa los recursos de manera más eficiente en el entrenamiento de una sola máquina, sino que también escala a miles de máquinas sin sacrificar la eficiencia de los datos o la utilización de recursos, lo cual se consigue al combinar la actuación y el aprendizaje desacoplados con un novedoso método de corrección fuera de la política llamado V-trace.

### 8.1.2.- Deep Actor-Critic

Las técnicas Actor-Critic son una versión de diferencia temporal (TD) del algoritmo Policy Gradients. Consiste en dos redes: el actor y la crítica. El actor decidía qué acción debía tomarse y el crítico le informaba qué tan buena era la acción y cómo debía ajustarse. El aprendizaje del actor se basa en el enfoque de gradiente de políticas. En comparación, la crítica evalúan la acción producida por el actor calculando la función de valor. Entre las técnicas modernas para implementar este tipo de modelos mediante aprendizaje profundo destacan las siguientes:
* **Asynchronous Advantage Actor-Critic (A3C)**: implementa capacitación paralela en la que varios trabajadores en entornos paralelos actualizan de forma independiente una función de valor global, por lo tanto, "asincrónica". Un beneficio clave de tener actores asincrónicos es la exploración efectiva y eficiente del espacio estatal.
* **Advantage Actor-Critic (A2C)**: es como A3C pero sin la parte asíncrona, funcionando para un solo trabajador. Se descubrió empíricamente que A2C produce un rendimiento comparable al de A3C y, al mismo tiempo, es más eficiente.
* **Soft actor-critic (SAC)**: es un algoritmo off-policy que forma un puente entre la optimización de políticas estocásticas y los enfoques de estilo DDPG. Incorpora el truco de la doble Q recortada y, debido a la estocasticidad inherente de la política en SAC, también termina beneficiándose de algo como el suavizado de la política de objetivos.

## 8.2.- Extensiones a los modelos

### 8.2.1.- Curriculum learning

A pesar de los muchos avances recientes, el aprendizaje por refuerzo en muchos dominios aún requiere una gran cantidad de interacción con el entorno, lo que puede ser prohibitivamente costoso en escenarios realistas.

Para abordar este problema, el aprendizaje por transferencia se ha aplicado al aprendizaje por refuerzo **empezando temporalmente por problemas más fáciles de resolver** e ir complicando su resolución poco a poco, de modo que la experiencia adquirida en una tarea se puede aprovechar al comenzar a aprender la siguiente tarea más difícil. Más recientemente, varias líneas de investigación han explorado cómo las tareas, o las muestras de datos en sí mismas, pueden secuenciarse en un plan de estudios con el fin de aprender un problema que, de otro modo, sería demasiado difícil de aprender desde cero.

### 8.2.2.- Imagination

Tal y como ocurría en métodos model-based reinforcement learning, a los agentes se les puede incorporar una nueva red con la que **predecir tanto los estados como las recompensas futuras** antes de llevar a cabo las interacciones con el entorno, lo que ayudará a que se tomen las decisiones óptimas de forma más rápida y efectiva (siempre y cuando las predicciones sean lo suficientemente acertadas en general).

Un algoritmo de este tipo es el **Imagination-Augmented Agents (I2A)**, que, a diferencia de la mayoría de los métodos de planificación y aprendizaje por refuerzo basados en modelos existentes, que prescriben cómo se debe usar un modelo para llegar a una política, los I2A aprenden a interpretar las predicciones de un modelo de entorno aprendido para construir planes implícitos de manera arbitraria, utilizando las predicciones. como contexto adicional en redes de políticas profundas.

### 8.2.3.-Neuroevolution

Este tipo de técnicas difieren en los demás principalmente en su método de entrenamiento, ya que no son entrenados mediante descenso por el gradiente sino mediante algoritmos evolutivos, generalmente algoritmos genéticos. El motivo de hacer esto es porque de este modo se pueden tratar de generar secuencias de estados y recompensas diferentes a los habituales, como por ejemplo un set de acciones de dimensión muy elevada, o que se quera definir una serie de recompensas multivariable más complejas o una estructura de red neuronal difícil de representar o entrenar mediante métodos tradicionales, por ejemplo.

Existen librerías python para entrenar este tipo de redes, como **Deep Neuroevolution** (basada en TensorFlow) o **PyTorch Neat** (basada en PyTorch).

### 8.2.4.- Multi-Agent RL

Multi-Agent Reinforcement Learning (MARL) es un subcampo del aprendizaje por refuerzo centrado en estudiar el comportamiento de **múltiples agentes coexistiendo en un entorno compartido**. Cada agente está motivado por sus propias recompensas y realiza acciones para promover sus propios intereses; en algunos ambientes estos intereses se oponen a los intereses de otros agentes, resultando en dinámicas grupales complejas.

El aprendizaje por refuerzo multiagente está estrechamente relacionado con la teoría de juegos y especialmente con los juegos repetidos, así como con los sistemas multiagente. Su estudio combina la búsqueda de algoritmos ideales que maximicen las recompensas con un conjunto de conceptos más sociológicos. Mientras que la investigación en el aprendizaje por refuerzo de un solo agente se ocupa de encontrar el algoritmo que obtiene la mayor cantidad de puntos para un agente, la investigación en el aprendizaje por refuerzo de múltiples agentes evalúa y cuantifica las métricas sociales, como la cooperación, la reciprocidad, equidad, influencia social, idioma y discriminación.

## 8.3.- Tecnologías para productivización

### 8.3.1.- Librerías de agentes

Aunque los modelos de DRL oueden ser implementados directamente en PyTorch o TensorFlow, es mucho más recomendable utilizar librerías de alto nivel en que se puedan utilizar modelos previamente desarrollados y optimizados. Entre las librerías a día de hoy más recomendables destacan las siguientes:

- **Stable Baselines**: librería desarrollada por OpenAI que implementa los principales modelos de DRL, siendo una evolución de OpenAI Baselines.
- **TensorFlow Agents**: librería desarrollada por Google enfocada a implementar los principales modelos de DRL en TensorFlow. 
- **Keras-RL**: librerías de agentes de DRL implementados en Keras.
- **Chainer**: una de las primeras librerías y más contrastadas de DRL.
- **RLLib**: librería de agentes DRL altamente paralelizable e industrializable.
- **TRFL**: (pronounced "truffle") is a library built on top of TensorFlow developed by DeepMind.

Existen otras muchas librerías, como PyQlearning, Tensorforce, RL Coach o Mushroom RL

### 8.3.2.- Librerías de entornos

Del mismo modo que podemos generar los modelos de redes neuronales que implementan las policies de los agentes, tammbién disponemos de múltiples entornos de trabajo en los que entrenar nuestros modelos y compararlos entre sí. Entre éstos destacan:

- **OpenAI Gym**: más utilizada librería con múltiples entornos de reinforcement learning.
- **Kaggle environments**: ofrecida por Kaggle para realizar competiciones entre usuarios.
- **Pyfolio**: utilizada para probar bots operando sobre mercados financieros.
- **PettingZoo**: especializada para entrenar multi-agent reinforcement learning.
- **PyBullet**: entornos pensado para robótica y disponibles en Google Colab.

Todos estos entornos disponen de una serie de métodos estándar para que sea replicables para múltiples modelos de agentes:
* <ins>Make</ins>: método con el que se invoca un determinado entorno de los existentes en la librería y se carga en memoria.
* <ins>Reset</ins>: se ejecuta cada vez que empieza un nuevo episodio, devolviendo al estado a su posición inicial y deshaciendo la memoria.
* <ins>Step</ins>: data una acción tomada ante un determinado estado, genera el estado siguiente, la recompensa de ese paso, un indicador que avisa de si el episodio ha terminado e información adicional customizable.

### 8.3.3.- Computación distribuida

Los modelos de DRL en muchas ocasiones dan problemas de convergencia o necesitan mucho tiempo de computación para que se comporten adecuadamente, según el problema. Por este motivo, durante los últimos años se han ido proponiendo diferentes técnicas para generar múltiples agentes en paralelo que, aunque compartan cierta información, vayan aprendiendo independientemente y mejorando la policy global de forma colaborativa. Algunas de estas técnicas son las siguientes:

* **Gorila**: presenta un marco general para paralelizar Q-learning profundo utilizando los componentes que describimos anteriormente. Una versión específica de esta arquitectura, implementada por los autores, agrupa a un actor, un alumno y un búfer de reproducción local para el aprendizaje.
* **Ape-X**: múltiples actores, cada uno con su propia instancia del entorno, genera su experiencia, agregarla a una memoria de reproducción de experiencia compartida y calcular las prioridades iniciales para los datos. El agente (único) toma muestras de esta memoria y actualiza la red y las prioridades de la experiencia. Las redes de los actores se actualizan periódicamente con los últimos parámetros de red neuronal.
* **Ray**: es un marco de ejecución distribuida de alto rendimiento dirigido a aplicaciones de aprendizaje de refuerzo y aprendizaje automático a gran escala. Logra escalabilidad y tolerancia a fallas abstrayendo el estado de control del sistema en un almacén de control global y manteniendo todos los demás componentes sin estado. En reinforcement learning se combina con la librería RLlib.

```python
import sys
sys.path.append("data_08/stocks/")

import torch
import pandas as pd
from modules.utils import resume
from modules.agent import rl_agent_train
from modules.environment import Environment
from tensorboardX import SummaryWriter
import yaml
import os
import yaml
from modules.model import Dueling_Q_Network
```

```python
def train(env, model, config):
    model.to(config['device'])
    model_ast = type(model)(model.input_size).to(config['device'])
    cuda = torch.cuda.is_available()

    if os.path.join("data_08/stocks/temp", config['resume_checkpoint']):
        if os.path.exists(os.path.join("data_08/stocks/temp", config['resume_checkpoint'])):
            _ = resume(
                model=model,
                cuda=cuda,
                resume_checkpoint=os.path.join("data_08/stocks/temp", config['resume_checkpoint'])
            )
        else:
            print('checkpoint: "{}" does not exist'.format(config['resume_checkpoint']))
            print('------------------------train from scratch------------------------------------')
    else:
        print('-----------------------------train from scratch------------------------------------')

    if torch.cuda.device_count() > 1:
        print(" let's use , {} GPU".format(torch.cuda.device_count()))
        model = torch.nn.DataParallel(model)
        model_ast = torch.nn.DataParallel(model_ast)

    model.train()
    model_ast.train(mode=False)
    optimizer = torch.optim.Adam(params=model.parameters(), lr=config['learning_rate'], weight_decay=0.0005)
    criterion = torch.nn.MSELoss()

    global_step = 0
    rewards = []
    losses = []

    writer = SummaryWriter(log_dir=os.path.join("data_08/stocks/temp" + '/tensorboard', config['checkpoint_dir']))
    rewards, losses = rl_agent_train(
        model=model,
        model_ast=model_ast,
        env=env,
        step_max=len(env.data)-1,
        epsilon=config['epsilon'],
        epsilon_min=config['epsilon_min'],
        epsilon_reduce=config['epsilon_reduce'],
        epsilon_reduce_freq=config['epsilon_reduce_freq'],
        device=config['device'],
        memory_size=config['memory_size'],
        global_step=global_step,
        train_freq=config['train_freq'],
        batch_size=config['batch_size'],
        discount_rate=config['discount_rate'],
        criterion=criterion,
        optimizer=optimizer,
        losses=losses,
        rewards=rewards,
        update_model_ast_freq=config['update_model_ast_freq'],
        checkpoint_dir=os.path.join("data_08/stocks/temp" + '/checkpoint', config['checkpoint_dir']),
        mode=config['mode'],
        writer=writer,
        save_freq=config['save_freq'],
        num_epoch=config['num_epoch']
    )

    return model, losses, rewards


def read_data(path, start_date, split_date):
    data = pd.read_csv(path)
    data['Date'] = pd.to_datetime(data['Date'])
    data = data.set_index('Date')

    if start_date:
        train_data = data[start_date:split_date]
    else:
        train_data = data[:split_date]
    test_data = data[split_date:]

    return train_data, test_data


def main(config):
    model = Dueling_Q_Network(181)
    train_data, test_data = read_data(
        path=os.path.join('data_08/stocks/ETFs/aadr.us.txt'),
        start_date=config['start_date'],
        split_date=config['split_date']
    )

    train_env = Environment(data=train_data, history_length=180)
    test_env = Environment(data=test_data, history_length=180)

    model, losses, rewards = train(
        env=train_env,
        model=model,
        config=config
    )

    return
```

```python
with open('data_08/stocks/config/config.yml', 'r') as stream:
    config = yaml.safe_load(stream)

print(config)
_ = main(config)
```
