## 9.1.- Visión general del NLP

### 9.1.1.- Aplicaciones empresariales

El Procesamiento del Lenguaje Natural (PLN) o Natural Language Processing (NLP) abarca todo el conjunto de técnicas orientadas a explotar información de tipo textual.

Mediante algoritmos de NLP es posible automatizar o al menos acelerar gran cantidad de actividades:
* <ins>Gestión de incidencias</ins>: dentro de las compañías se llevan a cabo interacciones tanto a nivel interno (con el servicio técnico, con recursos humanos, etc.) como externo por parte de los clientes. Mediante la aplicación de técnicas de NLP es posible interpretar muchas de estas interacciones y, en algunos casos, responder automáticamente, reduciendo las tareas de gestión en primer y segundo nivel.
* <ins>Análisis de conversaciones</ins>: generalmente las compañías llevan a cabo gran número de interacciones registradas en soporte digital con sus clientes, desde encuestas hasta e-mails o audios de conversaciones o intermediación con agentes. Todo esto puede ser supervisado y clasificado mediante técnicas de NLP.
* <ins>Análisis de marca en redes sociales</ins>: cada vez más datos sobre las marcas aparecen en internet, sobre todo en las redes sociales, por lo que muchas compañías precisan de sistemas capaces de rastrear gran cantidad de páginas web, incluyendo imágenes y vídeos, para interpretar y hacer un seguimiento de potenciales amenazas y oportunidades. 
* <ins>Sistemas de conversación</ins>: los sistemas de conversación poseen la ventaja de funcionar 24x7 y, aunque generalmente presentan problemas de falta de comprensión en la resolución de problemas más complejos, son capaces de filtrar una gran cantidad de incidencias sencillas o repetitivas para las que el chatbot responde de modo efectivo. Los sistemas de conversación en general se basan en un sistema de reglas, el cual se opera en función a un sistema de clasificación de intención y de reconocimiento de entidades, como veremos en detalle más adelante.
* <ins>Filtros de correo y ciber-seguridad</ins>: los algoritmos de NLP pueden usarse también, tanto como filtro de spam como detección de posible fishing  otras amenazas de ciber-seguridad. Obsérvese que las técnicas de NLP pueden operar después del procesamiento visual de textos en imágenes (OCR), de modo que todo lo que es legible para una persona también lo sería para un sistema cognitivo.
* <ins>Cognitive RPA / Cognitive Automation</ins>: cada vez se automatizan más actividades de bajo valor antes realizadas por humanos. En general se trata de actividades que conllevan acciones, que pueden automatizarse medianta las tecnologías RPA, pero también conlleva cierta capacidad de interpretación para, en función a ella, llevar a cabo una acción u otra. Este tipo de tomas de decisión basadas generalmente en visión (presencia o no de firma, tiempo desde fecha de facturación, etc.) o en análisis de textos (en algunos casos tras reconocimiento OCR o tras un Authomatic Speech Recognition), pertenecen al ámbito de los C-RPA o Cognitive RPA. 

### 9.1.2.- Técnicas tradicionales

En Natural Language Processing existen una serie de actividades genéricas que se repiten una y otra vez en cada una de las actividades antes mencionadas. Estas son las actividades más repetidas en la práctica:
- <ins>Eliminación de elementos irrelevantes</ins>: existen una serie de signos de puntuación que podrían ser eliminadas en un primer paso. También existen palabras con escaso o nulo valor para llevar a cabo la mayor parte de tareas de NLP, como artículos o muchos determinantes que suelen estar listadas en las librerías python como "**stopwords**" y que pueden ser eliinadas inicialmente para simplificar cálculos posteriores.
- <ins>Separación de tokens ("tokenización")</ins>: en cualquier lenguaje existen diferentes tipos de tiempos verbales, prefijos, sufijos, desinencias, etc. que aportan poco desde una perspectiva semántica al mismo tiempo que se necesita comprender gran número de expresiones regulares compuestas de varias palabras que conforman en realidad una unidad semántica que querríamos diferenciar. Mediante técnicas como la **lematización** (selección de la raíz léxica de cada palabra) y el **stemming** (eliminación prefijos y sufijos) se puede simplificar el vocabulario para que pueda comprenderse mejor el significado general de las frases. Existen también redes de palabras por significado como **wordnets**, que nos permiten identificar relacinoes semánticas entre palabras y su eventual sustitución cuando así se requiera.
- <ins>Análisis sintáctico</ins>: también existen librerías para identificar el **part-of-speech** de cada palabra del texto, pudiendo entender qué palabra es el verbo, cuál el sujeto, complemento directo, etc., lo que ayuda a la comprensión del texto.
- <ins>Representación / vectorización</ins>: para aplicar técnicas de machine learning, en general es necesario transformar el texto en vectores numéricos, lo cual se puede hacer de múltiples maneras: mediante conteos, mediante la técnica **TF-IDF** que destaca palabras características dentro de agrupaciones de sub-documentos, etc. Otra forma de representación muy utilizada es la representación en **embeddings como GloVe o Word2Vec**, que hoy en día se lleva a cabo mediante técnicas de aprendizaje profundo.
- <ins>Reducción dimensional</ins>: una vez el texto ha sido vectorizado, se pueden aplicar diferentes técnicas de reducción dimensional como SVD (Singular Value Decomposition), NMF (Non-negative Matrix Factorization) o PCA (Principal Components Analysis) para poder asociar documentos con los mismos rasgos. En las técnicas tradicionales se solía realizar la simplificación de considerar al texto una bolsa de palabras o **bag of words (BOW)** dentro de un párrafo, lo cual acarrea alguna imprecisión pero permite simplificar el problema para llevar a cabo tareas básicas y genéricas sobre el texto.
- <ins>Reconocimiento de entidades</ins>: conocidas también como **Name Entity Recognition (NER)** consiste en encontrar una serie de entidades dentro del texto, como fechas clave o lugares de interés, lo cual sirve para identificar información clave para alimentar look-up tables o incluso como entrada en grafos de conocimiento. Hoy en día esto se lleva a cabo casi exclusivamente mediante aprendizaje profundo, pero tradicionalmente se utilizaron con éxito modelos bayesiandos CRF o mediante los ya mencionados synsets de WordNets. 
- <ins>Clasificación de topics</ins>: esto se puede hacer de forma no supervisada (cuando no disponemos de etiquetas de topics en ninguna muestra) o supervisada (cuando sí que existen estas etiquetas sobre una muestra y queremos extenderla automáticamente a otros textos). Para el caso no supervisado, la técnica más utilizada es el **Latent Dirichlet Allocation (LDA)** y para las supervisadas, pueden servir desde redes **Naïve Bayes** como técnicas más avanzadas de machine learning como **SVM o Gradient Boosting** sobre la vectorización tras la reducción de dimensión.
- <ins>Detección de sentimiento</ins>: se desea interpretar si el bloque de texto posee un sentimiento positivo o negativo, lo cual se puede llevar a cabo sobre librerías basadas en reglas como **Vader** o mediante modelos de clasificación como los mencionados en el caso anterior. Nuevamente se trata de técnicas que hoy en día han sido casi completamente reemplazadas por técnicas de aprendizaje profundo.
- <ins>Reconocimiento de estilos</ins>: aparte de realizar clasificación de autores partiendo de diferentes textos partiendo de la simplificación BoW, también es posible entrenar modelos más sofisticados para aquellas tareas en las que el orden de las palabras jueguen un papel crucial, como en los **Hidden Markov Models (HMM)**, tanto sobre rañices y lemas de palabras como sobre estructuras gramaticales que pudieran caracterizar a diferentes autores.
- <ins>Sumarización</ins>: se trata de reducir un texto largo, reemplazándolo por otro que dice lo mismo pero es más corto. Nuevamente existen algoritmos para detectar las palabras y, sobre todo, frases clave, eliminando aquellas que aportan menos al texto completo. Se trata de otra actividad que funciona mucho mejor mediante técnicas de aprendizaje profundo.
- <ins>Traducción</ins>: se trata de obtener la misma información traducida a otros idiomas. Nuevamente las técnicas de aprendizaje profundo han alcanzado recientemente cotas impensables hace tan solo unos años, pero existen alternativas tradicionales basadas en librerías de diccionarios y traductores online a los que se puede invocar.
- <ins>Information Retrieval</ins>: se trata de una actividad que requiere en primer lugar de un sistema de interpretación de preguntas (**question-answering**), la cual ha de quedar vectorizada y comparada con la información previamente vectorizada de la documentación de búsqueda. Existen múltiples librerías para llevar a cabo este tipo de información de forma eficiente.

```python
# from https://www.kaggle.com/code/kunjan6902/doc2vec-word-embedding
```

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from nltk.tokenize import word_tokenize
from tqdm import tqdm
import numpy as np
from sklearn.manifold import TSNE
```

```python
# TRAIN
data = ["I love machine learning. Its awesome.",
        "I love coding in python",
        "I love building chatbots",
        "they chat amagingly well"]

tagged_data = [TaggedDocument(words=word_tokenize(_d.lower()), tags=[str(i)]) for i, _d in enumerate(data)]
```

```python
# doc2Vec model
max_epochs = 100
model = Doc2Vec(vector_size=20,
                epochs=10,
                alpha=0.025, 
                min_alpha=0.00025,
                min_count=1,
                dm =1)  

model.build_vocab(tagged_data)

for epoch in tqdm(range(max_epochs)):
    model.train(tagged_data,
                total_examples=model.corpus_count,
                epochs=model.epochs)
    model.alpha -= 0.0002
    model.min_alpha = model.alpha
```

```python
# GET EMBEDDING
def get_embedding(text):
    test_data = word_tokenize(text.lower())
    v1 = model.infer_vector(test_data)
    return v1

sentences = [
        "I love chatbots",
        "Machine learning is awesome",
        "Chatbots are amazing",
        "I love machine learning. Its awesome.",
        "I love coding in python",
        "I love building chatbots",
        "they chat amagingly well",
        "i do not like mondays!"
        ]
emb = []
for sent in sentences:
    v = get_embedding(sent)
    emb.append(v)
    print(sent,v)
```

```python
# T-SNE
X = np.array(emb)
S_t_sne = TSNE(n_components=2, init='random', perplexity=3).fit_transform(X)
    
# plotting scatter plot
x = np.array([v[0] for v in S_t_sne])
y = np.array([v[1] for v in S_t_sne])
plt.scatter(x, y)

for i,xy in enumerate(S_t_sne):
    print(xy,sentences[i])
    plt.text(xy[0],xy[1],sentences[i])
    
plt.show()
```

```python
import pandas as pd
import gensim
from gensim.utils import simple_preprocess
from gensim.parsing.preprocessing import STOPWORDS
from gensim import corpora, models
from nltk.stem import WordNetLemmatizer, SnowballStemmer
from nltk.stem.porter import *
import numpy as np
import nltk
nltk.download('wordnet')
np.random.seed(2018)
stemmer = PorterStemmer()
```

```python
# exploremos los datos
data = pd.read_csv('data_UD4/nlp1/abcnews-date-text.csv', error_bad_lines=False);
data_text = data[['headline_text']]
data_text['index'] = data_text.index
documents = data_text
print(len(documents))
print(documents[:10])
```

```python
id_text = 0
```

```python
doc_sample = documents[documents['index'] == id_text].values[0][0]  
print('original document: ')
words = []
for word in doc_sample.split(' '):
    words.append(word)
print(words)
```

```python
def lemmatize_stemming(text):
    return stemmer.stem(WordNetLemmatizer().lemmatize(text, pos='v'))

def preprocess(text):
    result = []
    for token in gensim.utils.simple_preprocess(text):
        if token not in gensim.parsing.preprocessing.STOPWORDS and len(token) > 3:
            result.append(lemmatize_stemming(token))
    return result

print('\ntokenized and lemmatized document: ')
print(preprocess(doc_sample))
```

```python
# Apliquemos esto a todos los textos
processed_docs = documents['headline_text'].map(preprocess)
processed_docs[:10]
```

```python
dictionary = gensim.corpora.Dictionary(processed_docs)
count = 0
for k, v in dictionary.iteritems():
    print(k, v)
    count += 1
    if count > 10:
        break
```

```python
dictionary.filter_extremes(no_below=15, no_above=0.5, keep_n=100000)
bow_corpus = [dictionary.doc2bow(doc) for doc in processed_docs]
bow_corpus[id_text]
```

```python
bow_doc_sel = bow_corpus[id_text]
for i in range(len(bow_doc_sel)):
    print("Word {} (\"{}\") appears {} time.".format(bow_doc_sel[i][0], 
                                               dictionary[bow_doc_sel[i][0]], 
bow_doc_sel[i][1]))
```

```python
tfidf = models.TfidfModel(bow_corpus)
corpus_tfidf = tfidf[bow_corpus]
from pprint import pprint
for doc in corpus_tfidf:
    pprint(doc)
    break
```

```python
lda_model = gensim.models.LdaMulticore(bow_corpus, num_topics=10, id2word=dictionary, passes=2, workers=2)
for idx, topic in lda_model.print_topics(-1):
    print('Topic: {} \nWords: {}'.format(idx, topic))
print(processed_docs[id_text])
for index, score in sorted(lda_model[bow_corpus[id_text]], key=lambda tup: -1*tup[1]):
    print("\nScore: {}\t \nTopic: {}".format(score, lda_model.print_topic(index, 10)))
```

```python
lda_model_tfidf = gensim.models.LdaMulticore(corpus_tfidf, num_topics=10, id2word=dictionary, passes=2, workers=4)
for idx, topic in lda_model_tfidf.print_topics(-1):
    print('Topic: {} Word: {}'.format(idx, topic))
print(processed_docs[id_text])
for index, score in sorted(lda_model_tfidf[bow_corpus[id_text]], key=lambda tup: -1*tup[1]):
    print("\nScore: {}\t \nTopic: {}".format(score, lda_model_tfidf.print_topic(index, 10)))
```

```python
unseen_document = 'How a Pentagon deal became an identity crisis for Google'
bow_vector = dictionary.doc2bow(preprocess(unseen_document))
for index, score in sorted(lda_model[bow_vector], key=lambda tup: -1*tup[1]):
    print("Score: {}\t Topic: {}".format(score, lda_model.print_topic(index, 5)))
```

```python
# import SentimentIntensityAnalyzer class
# from vaderSentiment.vaderSentiment module.
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
 
# function to print sentiments
# of the sentence.
def sentiment_scores(sentence):
 
    # Create a SentimentIntensityAnalyzer object.
    sid_obj = SentimentIntensityAnalyzer()
 
    # polarity_scores method of SentimentIntensityAnalyzer
    # object gives a sentiment dictionary.
    # which contains pos, neg, neu, and compound scores.
    sentiment_dict = sid_obj.polarity_scores(sentence)
     
    print("Overall sentiment dictionary is : ", sentiment_dict)
    print("sentence was rated as ", sentiment_dict['neg']*100, "% Negative")
    print("sentence was rated as ", sentiment_dict['neu']*100, "% Neutral")
    print("sentence was rated as ", sentiment_dict['pos']*100, "% Positive")
 
    print("Sentence Overall Rated As", end = " ")
 
    # decide sentiment as positive, negative and neutral
    if sentiment_dict['compound'] >= 0.05 :
        print("Positive")
 
    elif sentiment_dict['compound'] <= - 0.05 :
        print("Negative")
 
    else :
        print("Neutral")
 
 
# Driver code
if __name__ == "__main__" :
 
    print("\n1st statement :")
    sentence = "Geeks For Geeks is the best portal for \
                the computer science engineering students."
 
    # function calling
    sentiment_scores(sentence)
 
    print("\n2nd Statement :")
    sentence = "study is going on as usual"
    sentiment_scores(sentence)
 
    print("\n3rd Statement :")
    sentence = "I am very sad today."
    sentiment_scores(sentence)
```

```python
survey_text = (
    "Out of 5 people surveyed, James Robert,"
    " Julie Fuller and Benjamin Brooks like"
    " apples. Kelly Cox and Matthew Evans"
    " like oranges."
)


def replace_person_names(token):
    if token.ent_iob != 0 and token.ent_type_ == "PERSON":
        return "[REDACTED] "
    return token.text_with_ws


def redact_names(nlp_doc):
    with nlp_doc.retokenize() as retokenizer:
        for ent in nlp_doc.ents:
            retokenizer.merge(ent)
    tokens = map(replace_person_names, nlp_doc)
    return "".join(tokens)


survey_doc = nlp(survey_text)
print(redact_names(survey_doc))
```

## 9.3.- Grafos de conocimiento

### 9.3.1.- Representación del conocimiento

La representación del conocimiento y el razonamiento (**Knowledge Representation and Reasoning, KRR**) es el campo de la inteligencia artificial dedicado a representar información sobre el mundo en una forma que un sistema informático puede usar para resolver tareas complejas como diagnosticar una condición médica o tener un diálogo en un lenguaje natural.

La representación del conocimiento suele apoyarse en relaciones entre conceptos a través de **grafos de conocimiento** y en **ontologías**, definiciones formales de los tipos, taxonomías, propiedades, y relaciones entre entidades que realmente o fundamentalmente existen para un dominio de discurso en particular. Las ontologías contemporáneas comparten muchas similitudes estructurales, indiferente al lenguaje en el cual ellos fueron expresados. Los componentes más comunes de una ontología son:
- Individuos: instancias u objetos (lo básico u objetos de "bajo nivel" )
- Clases: conjuntos, colecciones, conceptos, clases en programación, tipos de objetos, o tipos de cosas.
- Atributos: aspectos, propiedades, rasgos, características, o parámetros que objetos (y clases) pueden tener.
- Relaciones: formas en las que las clases y los individuos pueden relacionarse entre sí.
- Funciones: Complejas estructuras formadas de cierta relación que pueden ser usadas en lugar de un término individual en una declaración
- Restricciones: establecen descripciones formales de lo que debe ser verdad con el objetivo de que alguna aserción pueda ser aceptada como entrada.
- Reglas: Declaraciones con forma de oraciones si-entonces (antecedente-consecuente) que describen inferencias lógicas que pueden ser derivables de una aserción en una forma particular .
- Axiomas: aserciones (incluyendo reglas) en una forma lógica que juntos incluyen toda la teoría que la ontología describe en su dominio de aplicación. Esta definición es diferente de los “axiomas” en gramáticas generadas y forma lógica. En esas disciplinas, axiomas solamente incluyen declaraciones especificadas como un conocimiento a priori. En las ontologías, "axiomas" también incluyen teorías derivadas de declaraciones axiomáticas.
- Eventos: los cambios de los atributos o relaciones.

Existen multitud de ontologías accesibles a través de internet, las cuales cubren todas las relaciones conceptuales y reglas lógicas dentro de un dominio de conocimiento especifico (medicina, ingeniería, lenguaje, etc.). Algunos de los grafos de conocimiento más conocidos son:
- **Google Knowledge Graph**: es una base de conocimiento estructurada en forma de grafo usada por Google para mejorar los resultados obtenidos con su motor de búsqueda mediante información de búsqueda semántica recolectada de una amplia gama de recursos.
- **DBPedia**: proyecto para la extracción de datos de Wikipedia para proponer una versión Web semántica. Este proyecto es realizado por la Universidad de Leipzig, Universidad Libre de Berlín y la compañía OpenLink Software.
- **Wikidata**: base de conocimientos gratuita y abierta que pueden leer y editar tanto humanos como máquinas, pues actúa como almacenamiento central para los datos estructurados de sus proyectos hermanos de Wikimedia, incluidos Wikipedia, Wikivoyage, Wiktionary, Wikisource y otros.

### 9.3.2.- WordNet

WordNet es una base de datos léxica que agrupa palabras en conjuntos de sinónimos llamados **synsets**, proporcionando definiciones cortas y generales y almacenando las relaciones semánticas entre los conjuntos de sinónimos. Su propósito es doble: producir una combinación de diccionario y tesauro cuyo uso sea más intuitivo, y soportar los análisis automáticos de texto y las aplicaciones de inteligencia artificial. Así, WordNet es el lexicón computacional más utilizado para desambiguar el significado de las palabras, una tarea que tiene como objetivo asignar el concepto más apropiado (vía synsets) a los términos en contexto. La base de datos y las herramientas del software se han liberado bajo una licencia BSD y pueden ser descargadas y usadas libremente a partir de la librería NLTK.

WordNet puede ser considerada un grafo de conocimiento, pero no como una ontología, pues la relación hiperónimo/hipónimo entre los synsets contiene cientos de inconsistencias semánticas básicas como la existencia de especializaciones comunes para categorías exclusivas y redundancias en la jerarquía de especialización.

### 9.3.2.- Web Semántica

La Web Semántica es una visión del futuro de la Web donde la información esta dando un significado explícito, permitiendo que las máquinas puedan procesar automáticamente e integrar la información disponible en la Web. 

El lenguaje de representación del conocimiento para crear ontologías dentro de la web semántica es Web Ontology Language (**OWL**), que se basa en el estándar del World Wide Web Consortium (W3C) para objetos llamado Resource Description Framework (**RDF**) y caracteriza por una semántica formal. ### 9.3.3.- Minikanren

El lenguaje miniKanren extiende el alcance de los sistemas anteriores para abordar problemas de lógica complejos a través de **programación lógica relacional**. Está basado en el lenguaje Scheme como host y es fácil de utilizar, pues sólo son indispensables tres operadores lógicos y un operador de interfaz. Se trata de un sistema que supera a lenguajes tradicionales de Inteligencia Artificial como Prolog y LISP en su **potencia y flexibilidad**, existiendo librerías para poder ser combinado con código Python.

```python
#!pip install miniKanren
```

```python
from kanren import Relation, facts, run, var, lall
parent = Relation()
facts(parent, ("Homer", "Bart"),
              ("Homer", "Lisa"),
              ("Abe",  "Homer"))
x = var()
```

```python
run(1, x, parent(x, "Bart"))
run(2, x, parent("Homer", x))

grandparent_lv, parent_lv = var(), var()
run(1, grandparent_lv, parent(grandparent_lv, parent_lv),
                       parent(parent_lv, 'Bart'))

def grandparent(x, z):
    y = var()
    return lall(parent(x, y), parent(y, z))

run(1, x, grandparent(x, 'Bart'))
```

## 9.4.- Sistemas conversacionales

### 9.4.1.- Fundamentos del sistema

Los sistemas conversacionales han sido ampliamente adoptados por muchas personas y organizaciones durante los últimos años. Permiten la comprensión de las necesidades del usuario, expresadas en lenguaje natural, y en el cumplimiento de dichas necesidades mediante la invocación de los servicios de back-end apropiados vía API.

En un sistema de conversación puede existir distintos componentes de tipo opcional, como los conversores Speech-to-Text o Text-to-Speech cuando se trata de comunicación oral, o traductores, correctores, etc., pero siempre existen dos componentes indispensables para el desarrollo de la conversación:
- <ins>State Tracker</ins>: es el componente encargado de descifrar y asimilar lo que el usuario ha dicho o querido decir, manteniendo un cierto contexto y memoria dentro de la conversación. Dentro del State Tracker los dos elementos más importantes de capturar son la **intención** del usuario mediante un modelo de clasificación, y las **entidades** o palabras clave mediante algoritmos de NER. Por ejemplo, un chatbot para reservar vuelos debe entender qué está preguntando el usuario (intención) y, una vez indique el vuelo que quiere, se extraiga la ciudad, la fecha, la edad de los pasajeros y otra información fundamental para la reserva. El State Tracker constituye de este modo un componente en el que el procesamiento del lenguaje natural es la pieza clave, aunque, al requerir cierto grado de formalización y sofisticación, siendo un híbrido entre sistemas KRR y NLP, se habla más a menudo de **NLU** (Natural Language Understanding). 
- <ins>Dialogue Manager</ins>: es el componente que se encarga del control del flujo de la conversación, siendo una de las tareas esenciales en los sistemas conversacionales y también la clave de su éxito y adopción. Aparte de **sistemas de reglas**, que son rápidos de implementar pero poseen muchas limitaciones, también se implementan **algoritmos de aprendizaje** supervisado (imitación o clonación de comportamiento) y por refuerzo (aprendiendo de las interacciones), desarrollándose de este modo sistemas conversacionales cada vez más complejos e indistinguibles del ser humano. El gestor de diálogo suele incorporar también un algoritmo de **generación de texto** de mayor o menor complejidad (desde listado de respuestas customizables a generación de texto de forma natural, para lo cual suelen requerirse redes neuronales profundas).

### 9.4.2.- Herramientas de desarrollo

Hoy en día existen gran cantidad de herramientas para desarrollar chatbots y todo tipo de sistemas conversacionales, de modo que ni siquiera es necesario ser un Data Scientist para implementar uno. Los más utilizados son los que se ofrecen de forma gratuita y los incorporados en los más importantes players de servicios cloud, como:
* **DialogFlow**: la propuesta de Google para desarrollar chatbots. Es fácilmente integrable en plataformas como Facebook, Twitter, Messenger, Whatssapp, etc. mediante un webhook, son fáciles de desarrollar y cubre la mayor parte de usos que podríamos necesitar. Sobre DialogFlow se puede poner un interfaz Google Assistant, pero igualmente se puede utilizar Alexa u otra tecnolgía.
* **AWS Chatbot**: basado en AWS Lex y altamente compatible con otras tecnologías como Alexa, dispone de un sistema fácil de utilizar similar a DialogFlow para desarrollar chatbots.
* **Azure Bot Service**: basado en el sistema NLP de Azure LUIS y altamente compatible con otras tecnologías, muy similar a los otros sistemas de chatbots de Amazon y Google.

Existen además otras alternativas del mercado como **Twilio**, orientado a la comunicación end-to-end, incluyendo la comunicación oral y sistemas de telefonía, para los que existen componentes que a manudo son utilizados también por chatbots desarrollados en otras tecnologías; así como otros muchos: Chatfuel, HubSpot, MobielMonkey, etc. También existe una herramienta para desarrollar chatbts más sofisticados y "out-of-the-box" sobre Python en los que nuestros conocimientos en procesamiento del lenguaje natural y sistemas de conversación marcan una diferencia significativa. En esta línea la librería más recomendable es **RASA**, una herramienta para crear chatbots de IA personalizados utilizando Python y la comprensión del lenguaje natural (NLU), lo que supone un paso más con respecto al NLP al añadirle comprensión de la estructura subyacente de intención/entidades y la posibilidad de implementar sobre él cierto razonamiento lógico basado en grafos de concimiento.
