## 1.2.- Contexto histórico del aprendizaje profundo

### 1.2.1.- ¿A dónde hemos llegado?

El aprendizaje profundo (Deep Learning) es una disciplina que ha gozado de grandes triunfos en los últimos años, llegándose a popularizar incluso a nivel popular, con éxitos conocidos como:

* **Generadores de imágenes** tipo Stable Diffusion, Midjourney o DALL-e
¿Pueden la inteligencia artificial generar arte? Sí, y en muy poco tiempo y con gran variedad de estilos. Este tipo de técnicas van a revolucionar el mundo del diseño y la creación, pues utilizan técnicas de aprendizaje profundo inventadas muy recientemente.

* **Motores de juegos** y vídeo-juegos como AlphaZero o AlphaGo
¿Pueden la inteligencia artificial ganarnos en todo tipo de juegos? Sí, y son capaces de vencer a los mejores campeones tanto humanos como cibernéticos. Este mismo de técnicas se están utilizando también para optimizar el comportamiento de sistemas mediante simuladores utilizando aprendizaje por refuerzo, lo que podría revolucionar la industria.

* **Chatbots** inteligentes a los que preguntar cualquier cosa: ChatGPT, basado en GPT-3.5
¿Puede la inteligencia artificial hacerse pasar opr una persona y contestarnos a cualquier tipo de preguntas sin que nos demos cuenta de que es una máquina? Sí, y puede revolucionar múltiples trabajos además de convertirse en un nuevo oráculo de conocimiento, reemplazando al buscador de Google.

Además de esto, simplemente en el año 2022 ha habido otras tantas iniciativas que se ha materializado en pocos meses: Whisper, GPT-3/4, Codex, GitHub Copilot, InstructGPT, Text-to-product, AI slides, Runway videos, Email AI, AI chrome extensions, Replit Ghostwriter, etc.

La preguntas que surgen son ¿en qué se basan estas técnicas? ¿cómo las podríamos utilizar?

Sobre todo, deberíamos empezar preguntándonos por entender este fenómeno desde sus orígenes ¿cómo hemos llegado hasta aquí?

La respuesta es que, aunque el aprendizaje profundo de redes neuronales ya existía hace más de 30 años, una confluencia de fenómenos ha hecho posible que experimentara el boom al que estamos asistiendo, siendo los principales factores de esta explosión los siguientes:

* El fenómeno **Big Data**, existiendo cada vez más datos digitalizados a partir del fenómeno web 2.0, tanto de datos numéricos estructurados como, sobre todo, datos no estructurados: texto, imágenes, audios, vídeos, etc. digitalizados y, en muchos casos, de libre acceso y subidos a la nube de forma distribuida por millones de usuarios.

* La proliferación del **código abierto** y una comunidad entusiasmada que comparte sus conocimientos a través de múltiples canales de comunicación, lo que ha permitido un proceso de aprendizaje colaborativo a nivel mundial como nunca antes ha existido en la historia.

### 1.2.2.- ¿Qué es el aprendizaje profundo?

El aprendizaje profundo (Deep Learning) hace referencia a las "redes neuronales profundas" (Deep Neural Networks), por lo que para entender qué es el aprendizaje profundo antes hemos de conocer lo que es una red neuronal artificial y en qué están basadas.

Las redes neuronales artificiales son modelos matemáticos que se componen de unidades interconectadas (neuronas) y conexiones en forma de pesos (dependencia entre ellas). En la práctica, se trata de funciones matemáticas en las que existe generalmente una capa de entrada, varias capas intermedias de neuronas (capas ocultas) y generalmente una capa de salida.

Como podemos ver en el diagrama de abajo, cada neurona recibe la entrada de varias neuronas conectadas a ella multiplicadas por un peso, agregadas posteriormente y transformadas por una función de activación. Repitiendo este proceso desde la capa de entrada hasta la de salida, se van obteniendo tanto los valores de las neuronas de las capas intermedias como de la capa de salida final. Este proceso se conoce con el nombre de **feed-forward o forward propagation**.

Como veremos a continuación, los primeros modelos de redes neuronales constaban apenas de dos o tres capas (**shallow neural networks**), pues se consideraba que con esta arquitectura ya se podían aproximar gran cantidad de funciones algebraicas. Se trataba de modelos en general bastante precisos pero difíciles de interpretar, de modo que su uso quedaba reducido a ciertos casos de negocio en los que la interpretación y el mantenimiento no era un problema y sí la precisión de las salidas. Esto hacía de las redes neuronales unos modelos que competían con otros de similar narturaleza como las Support Vector Machines o los ensemble de árboles de decisión (Adaboost, Gradient Boosting, Random Forest, etc.), que en muchas ocasiones daban similares o incluso mejores resultados que las redes neuronales. 

<img src="images/neural_network_example.png" alt="Alternative text" />

No fue hasta la década de 2010 cuando los modelos de redes neuronales tomaron una gran ventaja (debido a la confluencia de los tres fenómenos antes mencionados: Big Data, GPUs y código abierto), lo que permitió demostrar el enorme potencial de los modelos neuronales con múltiples capas ocultas (aprendizaje profundo) ya diseñados a finales del siglo XX por investigadores como Geoffrey Hinton, Yan Lecunn, Joshua Bengio y Carl Schmidhuber.

#### Tipologías dentro del aprendizaje profundo

El aprendizaje por refuero abarca tanto el aprendizaje **supervisado** (donde existe un target o ground truth al que queremos relacionar las entradas), como el **no supervisado** (sin la existencia de este ground truth) y el aprendizaje **por refuerzo** (interacciones basadas en recompensas). Se trata de sistemas de aprendizaje comunes al arsenal de técnicas dentro del machine learning. Aparte de estos tres sistemas de aprendizaje existen variates de los mismos, como el aprendizaje semi-supervisado, débilmente supervosado, auto-supervisado, etc.

El aprendizaje por refuerzo, como hemos visto, es sólo una parte del área de las **redes neuronales artificiales**, pues existen otros modelos de redes neuronales con pocas capas que no se consideran profundos pero sí redes neuronales. Igualmente, las redes neuronales son una técnica de machine learning, pero no la única (Existen otras técnicas como Gradient Boosting, K-medias, PCA, SVM, algoritmos evolutivos, etc. que pertenecen al machine learning pero no son redes neuronales).
Por último, el **machine learning**, tampoco abarcan toda la Inteligencia Artificial. Es más, es un avance bastante moderno pues, como veremos, los sistemas expertos soían estar entrenados directamente por árboles de reglas y grafos de conocimiento prefijados por humanos, de modo que no había realmente aprendizaje máquina. Sólo con la llegada de datos masivos de los que apender de forma automática empezó a tener verdadero peso el machine learning dentro del ámbito de la inteligencia artificial.

### 1.2.3.- Orígenes de las redes neuronales

#### Descubrimiento del funcionamiento de las neuronas y del orden sensorial

El primer paso adelante significativo hacia la comprensión del funcionamiento de nuestras neuronas lo dio precisamente un investigador español, <ins>Santiago Ramón y Cajal</ins>, el cual descubrió a finales del siglo XIX (premio Nobel en 1906) que el sistema nervioso consistía en una sucesión de células (neuronas), que **se comunicaban entre sí**, pero estando separadas físicamente unas de otras. Esa separación y, en conjunto, toda la zona que engloba el final del axón de una neurona, el espacio sináptico y el inicio de las dendritas de otra neurona se conoce como sinapsis. Es en la sinapsis donde se produce el salto de información de una neurona a otra a través de la liberación de neurotransmisores. Para desarrollar esta teoría revolucionaria, Ramón y Cajal se apoyó en el trabajo previo de investgadores como Luigi Galvani, que un siglo antes descubrió la existencia de actividad eléctrica en los animales, o Hermann von Helmholtz, que descubrió que la generación de electricidad por parte de los axones de las células nerviosas sería un medio para transmitir mensajes de un extremo a otro.

El trabajo de los pioneros de la neurociencia trascendió también al ámbito de la psicología. También en los albores del siglo XX se propuso el modelo **estímulo-respuesta** o de "aprendizaje por asociaciones" de <ins>Iván Pávlov</ins>, un tipo de aprendizaje asociativo inspirado en la ley de contigüidad de Aristóteles, que a su vez decía que cuando dos cosas suelen ocurrir juntas, la aparición de una traerá la otra a la mente. Parea demostrar su teoría, Pávlov ideó unos experimentos con perros que son la base del condicionamiento clásico: Se dio cuenta de que al ponerle la comida al perro, este salivaba. Cada vez que le pusiera la comida, Pávlov hacía sonar una campana, de modo que, cuando el perro la escuchaba, asociaba ese sonido con la comida y salivaba. Así, el perro estaba dando una respuesta (en este caso, la salivación) a un estímulo (la campana). La próxima vez que escuchara la campana, independientemente de si iba unida a la comida, empezaría a salivar. Este tipo de ideas servirían de inspiración no sólo a los modelos de redes neuronales, sino también a los sistemas de aprendizaje por refuerzo, basados en esta misma idea.

Durante los años 40 del siglo XX, <ins>Donald Hebb</ins> desarrollaría la primera teoría comprensible sobre el modo en que los fenómenos psicológicos tan complejos como las percepciones, las emociones, los pensamientos y la memoria, pueden ser producidos por la actividad cerebral. Su principal aporte se relaciona con la formación de ensambles neuronales regidos por el principio de que cuando el axón de una célula A está lo suficientemente cerca de una célula B como para excitarla y participa repetida o persistentemente en su disparo, ocurre algún **proceso de crecimiento** o cambio metabólico, en una o ambas células, de tal modo que la eficacia de A en disparar a B se ve aumentada. Este principio, en ciencia cognitiva, se denomina la "Regla de Hebb" y proveería el algoritmo básico de aprendizaje mediante redes neuronales artificiales.

Tal y como indicarían posteriormente los neuropsicólogos Joaquín Fuster y Steven Pinker, estas teorías llegaría a permear al ámbito de las ciencias sociales de la mano de <ins>F.A.Hayek</ins>, quien defendería durante los años 50 que la mente de los seres humanos, compuesta por neuronas interconectadas, va experimentando diferentes estímulos y procesos de aprendizaje que llevan a que se produzca lo que denominó como "**orden sensorial**", de modo que, a través de fenómenos a nivel microscópico (neuronas) emergerían comportamientos a escala macro no presentes en ninguno de ellos (acción humana), tal y como sucede también en el ámbito de la física de sistemas complejos. Utilizando esta idea como base de su Teoría del Conocimiento, Hayek sostendría la imposibilidad de acaparar todo el conocimiento existente en la sociedad por parte de una única unidad centralizada de conocimiento, de modo que el concimiento debería ser inevitablemente descentralizado y distribuido, pues sólo con la participación independiente de unidades de concimiento (las personas) y de instituciones sociales surgidas como estructuras emergentes del comportamiento humano (y no de forma centralizada, extrapolando de este modo la idea del orden sensorial al ámbito social, el "orden espontáneo") sería posible alcanzar un mayor estadío de concimiento orientado a la cooperación y, por tanto, de prosperidad dentro de la sociedad.

El conocimiento humano provendría, como sostendría más adelante <ins>Daniel Kahneman</ins>, tanto de procesos instintivos que ocurren en nuestra mente y que no somos capaces de intrepretar ("pensar deprisa"), en la línea del **conocimiento tácito** de <ins>Michael Polanyi</ins> (por ejemplo para aprender a montar en una bicicleta mediante puro instinto no verbalizable), así como de lo que nuestros antepasados erróneamente consideraban como única fuente de racionalidad ("pensar despacio"). La conducta resultaría de ambos procesos. No en vano, estas ideas, así como las ideas del modelo de estímulo-respuesta y del orden sensorial a través de la experiencia, servirían de inspiración para que <ins>Antonio Damasio</ins> formulara, ya a finales del siglo XX, su interesante teoría del **marcador somático**, la cual establecería que las emociones, no sólo el conoceimiento, proporcionan el andamiaje para la construcción de la cognición social y son necesarias para los procesos del yo que sustentan la conciencia, proporcionando una validación científica contemporánea del vínculo entre los sentimientos y el cuerpo al resaltar la conexión entre la mente y las células nerviosas, encarnación personalizada de la mente. Las ideas de Antonio Damasio estarían, de algún modo, haciendo desaparecer la frontera entre las emociones y el conocimiento, y quién sabe si también con ello se estaría diluyendo la frontera entre hombres y máquinas.

#### Primeros modelos de redes neuronales artificiales en el siglo XX

La más simple representación del funcionamiento de una neurona en términos de función matemática fue la **neurona de McCulloch-Pitts**, una unidad de cálculo que trataba modelizar el comportamiento de una neurona a semejanza de las que constituyen del cerebro humano. Este cálculo, desarrollado en 1940, consistía en la suma ponderada de las entradas, seguida de la aplicación de una función no lineal de tipo escalón, dando como salida la respuesta a a acumulación de impulsos.

Más adelante, en 1967, fue Frank Rosenblatt quien desarrolló y exploró todos los ingredientes básicos de los sistemas de aprendizaje profundo actuales: el perceptrón simple y el algoritmo de retro-propagación (backpropagation), el cual daría pie al desarrollo de modelos de **perceptrón multicapa**, así como el Neocognitron presentado por Kunihiko Fukushima en 1980, un antecedente de las redes convolucionales que se utilizarían para visión por ordenador unas décadas después. De hecho, En 1989, sería Yann LeCun quien aplicaría exitosamente el algoritmo de retro-propagación estándar, que había existido como el modo inverso de diferenciación automática desde 1970, a una red neuronal profunda con el propósito de reconocer códigos postales escritos a mano en el correo. Mientras el algoritmo funcionó, el entrenamiento requirió 3 días.

Durante la década de 1980 se fueron desarrollando diferentes modelos neuronales altermativos a los perceptrones multicapa, como las redes con **funciones de base radial** (RBF), que combinaban una etapa de clustering con otra de aprendizaje supervisado; o los **mapas autoorganizados** de Kohonen, un híbrido entre las técnicas de clustering y de reducción de dimensión; y las **redes de Hopfield**, con conexiones hacia delante y hacia atrás, orientadas al aprendizaje no supervisado y a la optimización. Estas últimas, además, darían pie a las primeras redes neuronales recurrentes pocos años después, las de Jordan y Elman, utilizadas por ejemplo para modelizar series temporales complejas, desde procesos químicos hasta la dinámica del habla.

<img src="images/som.png"  alt="Alternative text"  />

Sin embargo, a lo largo de la década de los 90s, nuevos modelos empezaron a restar atención de la comunidad científica hacia las redes neuronales, cayendo en lo que se conoce como el **invierno de las redes neuronales**. En especial, los modelos basados en Support Vector Machines (Vapnik, 1992) mostraron una eficacia superior en múltiples ámbitos a las redes neuronales, lo que unido a una mayor simplicidad en el entrenamiento que las redes neuronales más potentes, llevó a que los modelos de redes neuronales perdieran mucha atención entre mediados de los 1990 y la década de los 2000. 
Por ejemplo, la mayoría de los investigadores que trabajaban en modelos de reconocimiento de voz se alejaron de los modelos basados en redes neuronales para dedicarse al modelado generativo por medio de la combinación Hidden Markov Models con Gaussian Mixture Models (HMM-GMM), partiendo de coeﬁcientes cepstrales en las frecuencias de Mel (MFCC) como línea de desarrollo más prometedora en contraposición a los desarrollos basados en redes recurrentes llevados a cabo por investigadores como Hochreiter y Schmidhuber. Esta pugna se mantuvo desequilibrada hasta que estos últimos descubrieron una técnica para evitar el problema "vanishing gradients" de sus redes neuronales, desarrollando en 1997 el modelo LSTM, que en poco tiempo superaría a sus homólogos basados en HMM-GMM.

Otro gran avance se llevó a cabo en 2006, gracias a las publicaciones de Geoffrey Hinton, Ruslan Salakhutdinov, Osindero y Teh, en los que mostraron cómo una red neuronal feedforward de muchas capas podría pre-entrenarse capa por capa mediante técnicas de **pre-entrenamiento**, tratando cada capa a su vez como una máquina Boltzmann restringida no supervisada, y luego ajustándola usando retropropagación supervisada. Los investigadores estaban empezando a controlar la bestia de las redes neuronales profundas y todo un universo de aplicaciones se abrían a sus pies.

#### La revolución del aprendizaje profundo a partir de 2010

Fue a partir del año 2010 cuando los modelos de redes neuronales profundas empezaron a mostrar su enorme potencial. Los principales focos de desarrollo de este tipo de modelos estuvo en Canadá, especialmente en las universidades de Toronto (liderados por el británico **Geoffrey Hinton**) y Montreal (liderados por el francés **Yoshua Bengio**). Para ello aprovecharon inteligentemente el rápido desarrollo de la tecnología de tarjetas gráficas utilizadas en vídeo-juegos para poder entrenar complejos modelos como los diseñados por Yann LeCun, Geoffrey Hinton y Carl Schmidhuber años atrás.

Dentro del ámbito acedémico se consiguieron los primeros éxitos en 2011, obteniendo un rendimiento super-humano en un concurso de reconocimiento de patrones visuales y ganando el concurso de escritura a mano china ICDAR. Poco después, a principios de 2012, se ganó el concurso de segmentación de imágenes ISBI utilizando redes neuronales profundas y se realizaron sorprendentes contribuciones al ámbito del análisis de la actividad molecular. 

Sería, sin embargo en octubre de 2012 cuando las redes neuronales profundas trascenderían el ámbito académico y empezarían a sorprender al mundo gracias al arrollador trinufo en la famosa competición internacional **ImageNet-2012**. Esta competición supuso un antes y un después en el mundo del machine learning, pues el modelo AlexNet (llamado así por su creador Alex Krizhevsky, acompañado de Ilya Sutskever y Geoffrey Hinton) de redes neuronales profundas proporcionó tal salto cuántico en los resultados de los modelos de visión por computador, que al año siguiente casi todos los participantes se presentaron con modelos de aprendizaje profundo, mejorando aún más los resultados, y al año siguiente la práctica totalidad , alcanzando el nivel humano de percepción al tercer año, algo completamente impesable pocos años atrás.

<img src="images/imagenet.png" alt="Alternative text" />

El enorme éxito en la competición Imagenet impulsó la aplicación del aprendizaje profundo a otras muchas áreas, siendo el tratamiento de texto el otro campo en el que las redes neuronaes alcanzarían también un éxito sin precedentes e imaginable años atrás, como veremos en posteriores capítulos.

En 2014 se produjo otro hito importante sería la invención de los modelos generativos GAN (**Generative Adversarial Nertworks**) por el entonces joven estudiante de doctorado Ian Goodfellow, alumno de Andrew Ng (co-fundador de Google Brain) y Yoshua Bengio. Tras defender su tesis doctoral en 2014, toda una explosión de desarrollos en los modelos de generación de imágenes se abrió camino, siendo sus más populares hitos la aplicación FaceApp o como germen de los modelos de transformers que servirían de base a aplicaciones más modernas Stable Diffusion, Midjourney o DALL-e.

Sería en 2015 cuando el aprendizaje profundo volvería a sorprender al mundo. En este caso serían los modelos de aprendizaje por refuerzo profundo mediante Deep Q-networks para vencer en **juegos Atari** sin ningún tipo de instrucción ni recomendación, de modo que el sistema aprendía a superar la mayor parte de estos juegos por sí mismo a base de ensayo y error. Los investigadores que consiguieron este gran éxito estaban encabcezados por Volodymyr Mnih y David Silver, de la universidad de Toronto. En este grupo de investigadores estaba Demis Hassabis, fundador de la empresa DeepMind, el cual desarrollaría, con la ayuda de David Silver y su equipo, los más potentes motor de de ajedrez (AlphaZero) y del juego oriental go (AlphaGo) creados hasta la fecha.

El último gran evento notable vendría en el campo del procesamiento del lenguaje natural (Natural Language Processing, NLP). Durante esta década se conseguirían enormes avances en este campo aplicando modelos recurrentes y convolucionales, pero sería en 2017, con la aparición de los modelos de **Transformers**, cuando se daría un tremendo salto de calidad en este tipo de modelos. Los transformers permitirían el desarrollo de enormes modelos de aprendizaje profundo con los que representar la complejidad de lenguajes complejos, los cuales podrían ser reutilizados mediante transfer learning y fine-tuning y especialirlos en diferentes tareas. Este tipo de modelos, que se estudiarán más adelante, darían pie a los modelos BERT, GPT-x (1,2,3) y, más recientemente, al modelo ChatGPT, el cual combina modelos de transformers para el lenguaje (GPT-3.5) con modelos de aprendizaje por refuerzo en el gestor de diálogo tal y como descubrieron Mnih, Silver y Hassabis. Posteriormente estos modelos trascenderían el ámbito del procesamiento del lenguaje natural y actualmente son empleados en todo tipo de tareas (visión por computador, control de procesos, análisis de vídeo, etc.). Las principales compañías que están desarrollando este tipo de modelos son Google Brain, OpenIA y Hugging Face.

Todos estos investigadores pioneros gozarían de fulgurantes carreras a caballo entre la investigación dentro del ámbito universitario como en nuevas empresas de Inteligencia Artificial como OpenAI, DeepMind, Google Brain o Hugging Face, entre otras. En marzo de 2019, Yoshua Bengio, Geoffrey Hinton y Yann LeCun recibieron el **Premio Turing** por avances conceptuales y de ingeniería que han hecho de las redes neuronales profundas un componente crítico de la informática.

## 1.3.- Efectos económico-sociales del aprendizaje profundo

### 1.3.1.- Ampliación cognitiva

Hemos aprendido que, desde los albores de la humanidad, la evolución del cerebro humano ha sido fundamental para nuestra supervivencia y posterior dominio sobre las demás especies animales. No podemos saber cuál será nuestro futuro y si esta capacidad se convertirá algún día en nuestra perdición, pero lo que sí sabemos es que hasta ahora la prosperidad del ser humano ha estado ligada en gran parte a nuestra capacidad para tomer mejores decisiones y elaborar procedimientos (instituciones) y herramientas (bienes capital) que nos ayuden a alcanzar nuestros objetivos más consistentemente y con menor sacrificio. Y, sobre todo, nuestra capacidad cognitiva está muy ligada a facilitar nuestra capacidad de cooperación, pues el mayor uso de nuestro cerebro se orienta a comprender al más complejo y potente activo de los que nos rodea: otros seres humanos.

Hoy en día estamos viendo cómo el fenómeno del aprendizaje profundo, enmarcado de forma indivisible dentro de una revolución digital que se está acelerando durante los últimos años, está transformando nuestra forma de vida e impactando sobre la economía y la sociedad de nuestros días. El agente de cambio por el cual pasa toda esta transformación es el ser humano, que no sólo ve modificarse su entorno sino que experimenta lo que podríamos llamar una "ampliación cognitiva", pues de algún modo pasamos a ser canpaces de percibir mucha más información que antes no podíamos, siendo capaces de calcular más y mejor que antes, así como memorizar y crear, gracias a un soporte tecnológico a modo de superpoderes a nuestra disposición. Esta ampliación cognitiva se podría ilustrar del siguiente modo:
* **Percepción**: aunque con nuestros sentidos sólo podemos percibir lo que tenemos delante, hoy en día podemos ver a través de millones de cámaras por todo el mundo, drones que llegan a donde no podemos, por dentro de nuestro cuerpo mediante dispositivos médicos, acontecimientos pasados con gran lugo de detalle, etc. Todo este Big Data de esta <ins>visión extendida</ins>, además, es sintetizado por medio de algoritmos que nos ayuda a <ins>clasificar rápidamente</ins> lo que queremos dentro de este maremágnum y descartar lo que no. Veremos que en esta línea el aprendizaje profundo puede ayudarnos a analizar imágenes, audios, textos y todo tipo de información muy rápidamente para darnos directamente lo que queremos saber: <ins>dónde poner el foco</ins> y optimizar así nuestro tiempo. Hace que los médicos tengan ojos hasta en una aldea remota en el tercer mundo, que un sistema automático detecte anomalías en las cadenas de producción dentro de una fábrica sin estar presentes y que un chatbot nos encuentre rápidamente lo que queremos saber acerca de un cierto personaje histórico, por ejemplo. Nuestros sentidos se han extendido mucho más allá de nuestros límites físicos, por tanto.
* **Razonamiento**: a parte de extender nuestros sentidos mediante una mayor percepción, la IA del aprendizaje profundo también es muy potente para ayudarnos a razonar para tomar buenas decisiones. En primer lugar nos ayuda a predecir mejor a base de encontrar <ins>patrones causa-efecto</ins> que se repiten en el pasado, lo cual lo puede hacer mejor que nosotros, mucho más rápido y a una enorme escala. Esta mejor capacidad para predecir el futuro nos ayuda básicamente a <ins>anticipar</ins>, es decir, a tomar decisiones y llevar a cabo cambios para que el futuro que el sistema sea mejor para nosotros. Por ejemplo, hoy en día se utilizan masivamente modelos predictivos para mejorar las interacciones con los usuarios de una página web, por ejemplo, o para conocer los riesgos de crédito de una pesona física o jurídica, o para saber cuánta fruta se debe reponer en una tienda para que ni haya excedentes (lo que haría que se pudriera) ni que haya rotura de stock (lo que dejaría clientes insatisfechos). La mayor capacidad predictiva, por tanto, nos permite <ins>optimizar nuestras decisiones</ins>, lo cual nuevamente nos lo pueden resolver automáticamente las máquinas por medio de algoritmos metaheurísticos o sistemas avanzados de aprendizaje por refuerzo, como veremos más adelante.
* **Memoria**: nuestra memoria se desarrolló como mutación a lo largo de millones de años porque nuestros antepasados con mayor memoria repetían menos sus errores, tenía mayor éxito y sobrevivían más. Aunque nuestra memoria es limitada, la tecnología nos está proporcionando una memoria alternativa residente en <ins>discos duros</ins> de todo el mundo, alimentada además por experiencias de millones de personas o de sistemas autónomos. Superar las barreras físicas de nuestra memoria nos permite desarrollar programas cada vez más sofisticados en menos tiempo, pues no tenemos forzosamente que aprender toda la sintaxis y guardarla en nuestro cerebro, sino simplemente saber utilizar <ins>herramientas de búsqueda</ins>, alimentadas a su vez por inteligencia artificial, para reutilizar el conocimiento y la memoria de otras personas d modo rápido y eficaz. De este modo, la IA que ha propiciado el aprendizaje automático, especialmente el profundo, nos hace de brújula rapidísima dentro del Big Data, por ejemplo a través de técnicas como el neural search, el semantic hashing, information retrieval, etc. para interfaz cada vez más interactiva y conversacional. La memoria, además, permite en muchos casos <ins>reducir el riesgo moral</ins> apoyándose con la percepción, es decir, como se suele decir, "que paguen justos por pecadores", pues la información queda registrada y de este modo se desincentivan actos deshonestos. Este mejor sentido de responsabilidad es clave para incentivar la colaboración, como veremos más adelante.
* **Creatividad**: hace años era impensable que las máquinas, aparentemente aburridas y sin alma, nos pudieran ayudar en nuestra función creativa. Se ha demostrado, sin embargo, lo contrario, pues la creatividad humana se sustenta sobre una base de <ins>combinatoria de patrones</ins> y de matices relativamente sencilla que los sistemas neuronales captan y pueden replicar. Destaca aquí la <ins>creatividad compartida</ins>, donde el ser humano interactúa con el sistema de inteligencia en lugar únicamente de con su propio cerebro creativo, apoyándose en modelos neuronales generativos principalmente, como veremos más adelante. Hoy en día, y cada vez más, muchos trabajos creativos en el ámbito del diseño, literatura, pintura, música o incluso cine, son colaborativos entre personas y máquinas, de modo que el creativo puede mejorar mucho su productividad si aprende a utilizar estas tecnologías y se apoya en ellas. De hecho, la creatividad artificial de las máquinas es tal que nos ayuda también en actividades no artísticas a explorar nuevas alternativas. Es conocido el efecto del <ins>"movimiento 37"</ins> por una de las partidas de AlphaGO: el sistema de inteligencia artificial encontró mediante aprendizaje por refuerzo una jugada que jamás se le habría ocurrida a una persona, pero que desde entonces ha abierto nuevas formas de entender el juego muy interesantes y espectaculares. Es decir, los hombres estamos aprendiendo a partir de las nuevas ideas generadas por las máquinas. Es sorprendente ver cómo un sistema de IA puede jugar al ajedrez o al Go introduciendo nuevas y sorprendentes jugadas, siendo su juego nada aburrido, sino tremendamente arriesgado y creativo.

Por supuesto, esta ampliación cognitiva del ser humano impacta sobre todo el mundo e independientemente de sus intenciones. Igual que puede ayudar a buenas persona a mejorar la vida de los demás, también <ins>puede empoderar a malas personas</ins> que queran aprovecharse y hacer daño a otra gente. Lo importante en este caso es saber qué ventajas nos ofrece este fenómeno para, si nuestras intenciones son honestas, poder contribuir hacia lo que nosotros deseamos para nosotros y para la humanidad de forma pacífica y respetuosa.

### 1.3.2.- Procesos de auto-organización

Tal y como si se tratase de una reacción en cadena, esta ampliación cognitiva que nos proporciona la nueva inteligencia artificial impacta en segundo instancia sobre las **interacciones humanas** alterando nuestras relaciones. Como comentamos antes, los seres humanos hemos aprendido durante años a trabajar en equipo y en cooperar voluntariamente entre nosotros, bien sea mediante actividades remuneradas o no remuneradas. Ya desde los albores de la humanidad e incluso desde antes, la cooperación humana surge como mecanismo para mejorar nuestra supervivencia y lo hemos hecho, como no podría ser de otra manera, **por mera conveniencia personal**. A diferencia del resto de animales, sin embargo, nuestros mecanismos de cooperación son mucho más complejos y desarrollados gracias a una <ins>creciente capacidad cognitiva que la soporte</ins>, habiendose evolucionado a lo largo de millones de años instituciones sociales vitales para nosotros como el **lenguaje** y los **valores humanos**, sirviendo como mecanismos de cohesión social para favorecer nuestro **entendimiento mutuo** y el **trabajo en equipo**. Del mismo modo, incluso nuestros propios sentimientos están condicionados por el éxito de estas interacciones y la **empatía**, así como la producción de oxitocina y otros procesos biológicos que la propician, se ha consolidado del modo muy similar al que utilizan hoy en día las máquinas para aprender por refuerzo.

La nueva inteligencia artificial mejora nuestra capacidad para cooperar más exitosamente de múltiples formas, entre las que destacan:
* <ins>Detectando necesidades cruzadas</ins>: por ejemplo, mientras un particular podría estar buscando sin éxito los preparativos de un viaje, otro particular que dispone de dichos medios a buen precio desde otra parte del mundo, la tecnología puede ponerles en contacto para llegar a un acuerdo tal vez mutuamente beneficioso. La tecnología permite incluso que actividades muy particulares, incluso de tipo "long-tail" (minoritarias y difíciles de eoncotrar) encuentren su contraparte, pudiéndose producir el efecto "muñeca fea", mediante el cual algunos productos y servicios infravalorados dentro de una escala local pasen a ser muy valorados a escala global ante un mayor mercado que los valore por cómo son y los demande.
* <ins>Generando bienes de capital como excedente cognitivo</ins>: sin apenas coste para los particulares que los generan, se producen activos reutilizables y accesibles a terceros que impulsan el desarrollo de capital sin que esto suponga apenas coste ni esfuerzo adicional, sino como meras externalidades positivas del proceso productivo. Un ejemplo de esto es el desarrollo de código abierto, que está siendo crucial para el proceso en cadena de la inteligencia artificial, o hasta la mera aportación en un foro o red social se convierte en muchos casos en un activo reutilizable por terceros (sobre todo cuando se sabe hacer). Hoy en día cualquier puede reutilizar el conocimiento y la experiencia de terceros plasmada en forma de código reutilizable, librerías de libre acceso o recomendaciones en vídeos y foros abiertos de forma gratuita a todo el mundo con acceso a internet.
* <ins>Fomentando la competencia por la colaboración</ins>: a diferencia de lo que nos ocurría cuando apenas sabíamos cooperar entre nosotros y tendíamos a matarnos o en someternos los unos a los otros, en la actualidad hemos ido aprendiendo (aunque aún nos falta mucho) a trabajar en equipo por un mismo fin, compitendo pacíficamente y con reglas cada vez más justas, contra otros equipos que ofrecen lo mismo que nosotros. Esta competición es realmente muy importante para que aprendamos a coordinarnos eficazmente de forma descentralizada y a mejorar tanto nuestra tecnología como nuestros mecanismos de coordinación y distribución de trabajo. Este proceso de auto-organización se manifiesta, por ejemplo, entre empresas tecnológicas que desarrollan nuevos sistemas de inteligencia artificial y sin ella probablemente los logros actuales no habrían tenido lugar. Dentro de esta competición, siempre y cuando trabajemos con honestidad y reglas justas, nuestro trabajo se va revalorizando a medida en que aportamos nuevas ideas y ponemos nuestro granito de arena para el éxito del grupo, pues al mismo tiempo desarrollamos capacidades y experiencias que, aunque fracasemos eventualmente, contribuyen a que para próximos intentos aprendamos y tengamos más probabilidades de éxito, incluso en proyectos más aventurados.
* <ins>Mejorando la asignación de recursos</ins>: como decía Leila Janah, "el talento está igualmente distribuido, pero las oportunidades, no". Lo mismo ocurre con el esfuerzo, pues vivimos en un mundo con profundas desigualdades en el que algunos con muy poco esfuerzo pueden aspirar a una mejor vida que otros dejándose la piel. No obstante, la tecnología actual también puede contribuir a que esto continúe cambiando. En primer lugar, ayudando a que la tierra "se aplane" y desaparezcan estas barreras para el acceso a este nuevo activo mayormente gratuito. Hoy en día la educación es mucho más accesible a cualquier persona y acaba llegando a particulares con talento y capacidad de sacrificio, pudiéndoles servir como escalera hacia un trabajo que les acerque a una vida más próspera, como nunca antes en la historia. Estos mecanismos naturales de asignación de recursos que se van extendiendo no sólo sirven para que el mundo sea más justo con los que más aportan, sino que también producen efectos de segunda ronda consiguiendo que personas más capaces y trabajadoras generan más riqueza para los demás. La inteligencia artificial hace posible que no sólo estas tecnologías que sirven de base a estos procesos funcionen mejor, sino que también generan efectos similares a menor escala. Por ejemplo, una mejor capacidad para optimizar el inventario de productos perecederos hace que se desperdicie menos producto y que se perda menos tiempo en comprarlos al haber menos roturas de stock, o un buen sistema de captación de riesgos sirve para que el crédito llegue a quien va a hacer mejor uso de él o que las máquinas que automáticamente gestionan la ventilación de un metro consuman menos energía maximizando el confort.

En definitiva, el modo en que la esta ampliación cognitiva influye en mejorar el mundo (también en empeorarlo si viene utilizado por malas manos) pasa por una mejor capacidad para trabajar en equipo de forma completamente descentralizada, siendo capaces de encontrar nuestro sitio en el mundo aportando lo que otros necesitan y <ins>darnos cuenta de la mejor manera de aportar nuestro granito de arena al conjunto de necesidades globales</ins>. No se trata de un proceso centralizado y fácilmente concebible ni siquiera por el mejor sistema de inteligencia artificial, sino que opera <ins>a pequeña escala y de forma descentralizada</ins>, pacífica y respetuosa con las preferencias de los demás.

### 1.3.3.- Desarrollo económico

Hemos visto cómo la nueva inteligencia artificial nos ayuda a ampliar nuestro conocimiento y, con ello, nos empodera para poder llevar a cabo proyectos empresariales, remunerados o no, que en general mejoran la vida de los participantes. La tercera derivada consiste en pasar de este ámbito micro de ejemplo concretos a un ámbito macro tanto a escala mundial como a medio y largo plazo. ¿Cómo podría afectarnos el desarrollo de estas tecnologías al conjunto de la humanidad? ¿mejorará la vida de la gente? ¿cómo podríamos conseguir que así fuera?

Ante esta pregunta, muy difícil de contestar, sólo podemos decir que excede nuestra capacidad cognitiva. Tampoco ninguna inteligencia artificial nos lo va a decir porque a este tipo de juegos tan abiertos y complejos no saben jugar ni tienen de dónde partir. Lo más que podemos decir es que nuestra participación debe ir orientada a producir **efectos positivos a pequeña escala** aprendiendo de nuestros errores y operando con responsabilidad y ética a raíz de los resultados de nuestras acciones, efectos que en general no son del todo predecibles. Querer llevar a cabo grandes cambios, en general al margen de los deseos de los demás y sin respetar su voluntad, en general lleva al fracaso, por muy buenas intenciones que creamos que tenemos. En cambio, si nos centramos en aportar a la sociedad desde un marco de consentimiento y acptación mutua, algunas contribuciones a pequeña escala podría ser:
* <ins>Ayudando a desarrollar nuevas tecnologías</ins> que mejoran la vida de las demás personas, especialmente de las que más lo necesitan. Por ejemplo, desarrollar un sistema de inteligencia artificial que sepa interpretar las ondas cerebrales de pacientes que no pueden comunicarse o descubriendo cualquier mecanismo que ayude a derrotar a alguna enfermedad. También desarrollando sistemas que sirvan de base a estas tecnologías, como algoritmos de aceleración de búsquedas o de ejecución más eficiente contribuimos indirectamente.
* <ins>Creando puestos de trabajo</ins>: en muchas ocasiones lo mejor que podemos hacer para combatir la pobreza no es dar el pez a quien lo necesita, sino confiar en ellos y darles una oportunidad de formar equipo y juntos producir riqueza. También nuestra experiencia puede ayudar cuando la tengamos, pues eso les hará mejores profesionales y que su capacidad de generar riqueza y valor de mercado se incremente.
* <ins>Proporcionando financiación</ins>: cuando las dos alternativas no sean posible porque nuestras capacidades y oportunidades a nuestro alcance no den para tanto o no nos lo permitan, siempre nos queda trabajar en proyectos que, aunque no tengan un impacto directo en la vida de los más necesitados, nos proporcionen una renta que nos permita atesorar y ahorrar. Esto también es muy importante porque en muchos casos lo más necesario para combatir la pobreza es financiación, pero no cualquier financiación sino financiación inteligente y bien gestionada. Hoy en día existen medios para que podamos comparar diferentes proyectos humanitarios o incluso es posible financiar directamente proyectos de personas emprendedoras sin recursos. Se trata de escoger cómo distribuir nuestros recursos teniendo en mente que un proyecto de inversión llevado a cabo por una persona responsable vale por dos, pues si son capaces de devolver lo prestado, el crédito seguirá fluyendo hacia otros que lo necesiten y mejorará a largo plazo la vida de estas personas y de su entorno.

En definitiva, somos nosotros los agentes de cambio y todo lo que suceda dependerá al menos parcialmente de cada uno de nosotros. Si sabemos sacar partido a toda esta tecnología naciente tal vez podamos alcanzar nuestros fines materiales, satisfacer las necesidades de nuestra familia o la de la gran familia de la humanidad y de sus más necesitados. Puede aprendamos a producir cada vez más con menos esfuerzo, ampliando nuestro tiempo libre y ocio, o puede que no queramos ahorrar ni trabajar y que nada cambie, sólo se mantenga. Será como surcar las olas de un navío en una tempestad, habrá que ir improvisando, pero desde luego <ins>cuantos más recursos, conocimientos y experiencias tengamos</ins>, más probabilidades de éxito tendremos.
