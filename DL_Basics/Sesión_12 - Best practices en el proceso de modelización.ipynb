## 13.1.- Fase de desarrollo del modelo

### 13.1.1.- Propuesta y plan de proyecto

Durante estos últimos años, muchas compañías han empezado a emprender proyectos que involucran el desarrollo y/o la implantación de sistemas de inteligencia artificial, muchos de los cuales basados en modelos aprendizaje profundo. Al estar ante una tecnología que aún no ha madurado lo suficiente como para llegar a estar consolidada dentro de las organizaciones, lo que más se suele demandar por parte de las empresas son pruebas de concepto (PoC) en la que proveedor y cliente deben trabajar conjuntamente para dar respuesta a lo siguiente:
- Si es <ins>viable de escalar y rentable implantar</ins> dicha tecnología en la empresa o si aparece algún impedimento que habría que valorar si es posible y, en definitiva, conveniente superar.
- Si la tecnología <ins>funciona para las casuística de la empresa</ins> y dentro de la operativa normal, sin interferir en procesos clave de la compañía y de forma alineada con los procesos de cambio de la empresa y sus objetivos a distintos plazos.
- Si existen los <ins>medios y la capacidad suficiente</ins> como llevarla a cabo con éxito al menos en un medio plazo, debiendo esperarse rendimientos con el suficiente retorno de la inversión en un plazo de tiempo asequible y alineada con los objetivos y prioridades de la situación de la compañía.

A la hora de llevar a cabo estas pruebas de concepto es muy importante tener claro quién es el patrocinador del proyecto y cuáles son sus prioridades. Es importante que el equipo de cliente que se designe para el proyecto esté involucrado y alineado con los <ins>objetivos del patrocinador</ins> para evitar desalineamientos de expectativas.

Si la prueba de concepto es exitosa, el proveedor debe proponer un plan a medio-largo plazo para conseguir resultados productivizables en un horizonte cercano, calculando el retorno de la inversión y el payback del proyecto. Para ello se debe entender en qué fase de implantación de la tecnología se está, pues suele ser de gran utilidad fijar una primera fase en la que únicamente las compañías se centren en el desarrollo de un producto mínimo viable (**Minimum Viable Product, MVP**) con las funcionalidades reducidas al máximo, para posteriormente, generalmente a través de <ins>sprints agile</ins>, se vayan agregando funcionalidades hasta llegar a la implantación completa.

De cara a la oferta económica, se debe analizar la situación, la importancia  estratégica del cliente, expectativas de crecimiento, etc. para valorar cuál de las siguientes alternativas nos conviene apostar:
- **Proyecto a precio cerrado**: se calculan los costes esperados, se añade el margen de mercado y se valora si como cliente se ha de invertir en el proyecto, bien sea vía coste o vía expectativas de tiempo de trabajo no facturado. Al contratarse un producto, el proyecto se cerrará cuando este producto se haya terminado y entregado en forma, por lo que cualquier retraso no justificado caerá de parte del proveedor.
- **Time and materials**: no se contrata el producto ni los resultados, sino las horas/hombre. A menudo es la alternativa preferida para proyectos de pequeña escala con bajo presupuesto o de baja duración, siempre y cuando se dispongan de los perfiles adecuados para llevar a cabo el proyecto y, además, se disponga de cierto expertise dentro de la empresa cliente para supervisar los resultados obtenidos por el equipo del proveedor.
- **Proyecto value-based**: se proponen diferentes métricas de éxito y cómo calcularlas para, a continuación, proponer al cliente que la facturación no dependerá tanto de los costes internos de desarrollo, sino del impacto conseguido a través de estas métricas. Esta modelidad descarga gran parte del riesgo sobre el proveedor, por lo que sólo se debe hacer en condiciones especiales y donde se tenga una alta seguridad de éxito en el proyecto.
- **Servicio recurrente**: otra alternativa de interés consiste en acordar con el cliente un período mínimo de explotación de un determinado servicio recurrente. Si el cliente garantiza la contratación de dicho servicio siempre y cuando alcance unos estándares de calidad convenidos, el proveedor puede estar las primeras semanas o meses a fondo perdido porque tiene la certeza de que dispondrá de un ingreso recurrente una vez este haya sido completamente implementado. Se trata de una alternativa cómoda para el cliente, pero que sólo se puede llegar a acordar en caso de que sea una actividad consolidada o en la que ya se tenga cierta experiencia.
- **Licencias y propiedad intelectural**: otra forma de cerrar un acuerdo de colaboración en materia de innovación es establecer quién poseerá la propiedad intelectual del producto resultante (pueden ser ambos a un cierto porcentaje) para ofrecerlo y explotarlo con terceros. Esta opción entra en consideración cuando se está desarrollando un producto o servicio que podría ser de interés a terceros con los que se tienen contacto, además de alto potencial de crecimiento, lo que permitiría recuperar con creces la inversión del desarrollo inicial.


### 12.1.2.- Preparación del entorno de trabajo

A la hora de comenzar a trabajar en un proyecto, se debería previamente haber establecido con qué medios se trabajará en el proyecto y bajo qué condiciones de privacidad y confidencialidad. En general, cliente y proveedor suelen firmar un documento **Non Disclosure Agreement (NDA)** en el cual ambas partes se comprometen a cumplir ciertas reglas de confidencialidad y seguridad del dato, tras lo cual ya se puede compartir la información establecida bajo los términos acordados. En general, se deben rechazar aquellos datos adicionales que nos hubieran sido entregado sin pedirlos, sobre todo si contienen información confidencial. En caso de que accidentalmente se reciban datos de alto valor pero no acordados, se debe hacer una petición formal de ello y, sólo si se llega a un acuerdo preferiblemente por escrito, se deberán utilizar.

Una vez que ya se tiene acceso a los datos y nos disponemos a empezar con el tratamiento del dato, lo siguiente es pprepaar el entorno de trabajo, instalando las librerías que necesitemos, una vez hayan sido éstas aprobadas por el propietario del host.
En Python, en concreto, suele ser conveniente crear un **entorno virtual** de trabajo, bien sea con conda/miniconda o con virtualenv. Hacer esto y generar un fichero de requirements ajustado nos evitará problemas de compatibilidad entre el nuevo proyecto y otros que tuviéramos desplegados en dicha máquina.

A la hora de desarrollar el modelo, también conviene utilizar herramientas de monitorización y control de las distintas fases de la modelización, más aún si esto nos sirve para poder llevar a cab en el futuro actividades de DevOps. Este tracking se puede llevar a cabo mediante herramientas como **MLFlow, Comet o Kubeflow**. En todo caso, el código debe estar debidamente comentado y ser "**pythonic**", es decir, seguir los estándares habituales de modelización de Python para evitar errores, que esté optimizado y que sea facil de mantener. Por último, se debe dejar espacio para ficheros temporales, logs del proceso y carpetas de profiling y tests unitarios. Es más que recomendable utilizar **repositorios de código** como Git, siguiendo también sus best practices (utilizar branches, proponer merge requests al main, sólo desplegar código debidamente testado con reviewers, etc.) para evitar perder el trabajo de los días anteriores y no generar conflictos durante el trabajo colaborativo.

### 13.1.3.- Etiquetado de datos

Los modelos de aprendizaje profundo normalmente tienen el problema de que necesitan **gran cantidad de datos** para que den buenos resultados. Esto es, en parte, un problema, pero, por otro lado, saca a relucir una gran virtud de este tipo de modelos: su capacidad para seguir aprovechando detalles en los datos más allá que otras técnicas. Algunos modelos, como los basados en transformers, requieren tal cantidad de datos y poder de computación para alcanzar su nivel de precisión que conocemos, que sólo grandes compañías pueden asumir dicho esfuerzo. La ventaja de estos modelos es que, a diferencia de otros modelos de aprendizaje profundo, no sólo se comparte la estructura del modelo, sino los propios pesos del mismo, de modo que a los usuarios sólo nos queda llevar a cabo tareas de **fine-tuning** mucho menos costosas en términos de tiempo y esfuerzo.

El esfuerzo de etiquetado, es decir de construir los valores de "ground truth" que sirvan de supervisión, bien sean etiquetas binarias, categóricas, valores numéricos o conjuntos de pixels, hace que los modelos no supervisados o levemente supervisados (weakly-supervised o semi-supervised) permitan omitir o reducir este paso a vaces tan costoso. Afortunadamente también existen técnicas para maximizar este proceso de etiquetado, como el **active learning**, en el que el equipo de personas encargados de generar las etiquetas pueden apoyarse en algoritmos matemáticos que pre-calculen la diversidad (ofreciendo imágenes o textos muy variados al humano) y la incertidumbre (ofreciendo las imágenes o textos más ambiguos de cara a la clasificación) de cara a reducir los tiempos de etiquetado y que el equipo de etiquetadores se centre únicamente en los casos difíciles mientras que la Inteligencia Artificial automatice el resto.

A la hora de etiquetar los datos es preciso tener muy en cuenta la **privacidad**, sobre todo si se trata de datos personales, por lo que se debe asegurar que estén debidamente protegidos durante todo el pipeline por el que pasen, desde su captación a su explotación final. En algunos casos se pueden llegar a generar nuevos datos utilizando modelos generativos, lo que también ayuda a mantener la privacidad.

### 13.1.4.- Procesamiento de datos

Los modelos de aprendizaje profundo son famosos por requerir de poco trabajo de **pre-procesamiento** antes del entrenamiento. Esto sin duda es una ventaja en la mayoría de los coasos, pero también existe el riesgo de que dificultemos el cálculo del modelo por no tomar ciertas precauciones y llevar a cabo algún pre-procesamiento sencillo que facilite después la convergencia del modelo durante el entrenamiento, como eliminar aquellos que no posean un mínimo de calidad, normalizar algunos valores, filtrar en un paso más las regiones de interés o muestrear aleatoriamente los datos para evitar sesgos. También suele ocurrir que dispongamos de **información incompleta (missings)** que de algún modo tengamos que completar manualmente o mediante algún automatismo para que el modelo pueda hacer su función o eliminar **valores extremos (outliers)** a la hora de definir las recompensas en modelos de aprendizaje por refuerzo, por ejemplo.

### 13.1.5.- Entrenamiento y validación

Una vez ya tenemos los inputs y los outputs de nuestro modelo, llega el momento de definir la arquitectura del mismo, su función de pérdidas, criterio de optimización e hiper-parámetros para empezar con el entrenamiento. Es conveniente que, de cara a dejar el proceso iterando, llevemos a cabo una serie de **pruebas** para ver si las métricas de ajuste que definamos (accuracy, mae, mape, precision, recall, f1-score, ROC-AUC -Area Under Curve-, etc.) están resultando prometedores tanto en train como en test. Ya sabemos que si el modelo muestra buenos resultados en train, pero no en test, es que necesita mejores técnicas de regularización y que, si no muestra mejora en ninguno, es probable que haya algún error o que, sencillamente, no exista correlación o falten datos de calidad para poder entrenar. Una vez que hayamos hecho una serie de pruebas, bien automáticamente o manualmente, ya podemos dejar el proceso en ejecución por un tiempo y llevar a cabo otras tareas mientras tanto.

La **gestión del tiempo** en estos casos es fundamental, pues es una buena práctica aprovechar las noches o las horas de la comida para aprovechar el tiempo al máximo, siempre y cuando nos hayamos asegurado de que el modelo no va a dar algún error mediante unit testing o que no esté llevando a cabo correctamente el proceso de entrenamiento.

La validación final es, en última instancia, el cumplimiento de los objetivos del proyecto. No es necesariamente la precisión, aunque habitualmente existe correlación. Es una buena práctica asegurarnos de que las métricas que hayamos escogido para comparar modelos y escoger la variante definitiva está al menos alineada con la utilidad del modelo de cara a la resolución de los objetivos del proyecto, lo que significa que variables como **facilidad de mantenimiento**, **coste computacional** o **facilidad de explotación**, entre otras, también participan en la ecuación y han de tenerse muy en cuenta en todo momento.

## 13.2.- Productivización de los modelos

### 13.2.1.- Despliegue como servicio

En general, los modelos suelen ir desplegados dentro de alguno de los pipelines en producción, bien sea en tiempo real o en batch, pero cada vez más queda dentro de las responsabilidades de los Data Scientist su **encapsulado y mantenimiento**, o al menos planificarlo bien. Es decir, incluso cuando no nos corresponda a nosotros desplegar el modelo porque lo lleven a cabo otros equipos, debemos entregar a nuestros compañeros una **documentación completa y bien desarrollada**, de modo que contenga toda la información que sabemos que necesitarían para poder llevar a cabo su trabajo: datos de entrada, dependencias, memoria consumida, disco duro, consumo de procesador, end-points y puertos utilizados, etc. con su debida documentación técnica para que puedan monitorizar y mantener los modelos sin tener que concer en detalle qué hay por debajo.

De cara a encapsular y mantener el modelo, debemos considerar si nuestro modelo debería ir desplegado como un servicio al que se invoca a demanda o si simplemente debería formar parte de algún proceso en batch. En ambos casos es conveniente que se trate de encapsular todo lo que hagamos de modo que se replique de forma exacta el **entorno** con librerías python que en su momento utilizamos, las cuales deberán estar debidamente incluidas en el fichero **requirements.txt** incluyendo su número de versión exacta que nos está funcionando para la versión de Python que coresponda. En caso de entregar un contenedor docker, que también es una buena práctica, todo esto irá incluido dentro del **dockerfile**, de modo que nuestros colegas nos lo agradecerán.

También puede darse el caso de que seamos nosotros los que debamos **desplegar el servicio**, para lo cual existen herramientas muy alcance de los Data Scientists, como Flask/Django (síncronos) o Quart/FastAPI (asíncronos). Recientemente las librerías asíncronas han mostrado sus ventajas y se están empezando a imponer en nuevos proyectos, pero las anteriores siguen siendo un estándar para muchos proyectos. En el caso de utilizar Flask, se necesita que se incluya dentro de un software de servidor que gestione el balanceo de carga, la seguridad (como asegurar la comunicación por https) y las instancias, entre otras cosas, lo cual se suele hacer habitualmente por medio de la dupla NGINX/uWSGI y docker compose. Es preciso tener en cuanta las posibles interferencias entre sesiones y tartar de optimizar memoria haciendo que el servicio sea en lo posible de tipo stateless y facilitar su explotación y documentación por medio de flasgger.

### 13.2.2.- Desarrollo de interfaces de usuario

Nuevamente podríamos estar hablando de otro caso de "intrusismo" de los Data Scientist en el trabajo de los desarrolladores de front-end, pero de cara al prototipado y pruebas de concepto es una actividad que en numerosas ocasiones recae de nuestro tejado, pues entendemos mejor lo que necesitamos, cómo lo queremos y no necesitamos que el aspecto y su funcionamiento interno esté optimizado para cualquier Mproducto mínimamente viable.

Afortunadamente, existen librerías python para desarrollar interfaces de forma muy sencilla, tanto para ser instalados en local (Tkinter, Kivy, PyQt5, wxPython, etc.) como a través del navegador sin necesidad de instalación, tales como:
* **Dash**: en combinación con la librería gráfica plotly ofrecen un interfaz desplegable con diferentes botones interactivos y gráficos actualizables en tiempo real con los que cubrir gran parte de funcionalidades que pudiéramos requerir.
* **Streamlit**: a diferencia de dash, los componentes son más fácilmente customizables y muy fáciles de disponer en la pantalla. Se trata de un servicio accesible desde el browser y montado sobre el web server Tornado.
* **Gradio**: nueva librería de generación de interfaces todavía más fácil de utilizar, adecuada para desplegar servicios básicos de interacción con imágenes y textos a través del browser.

Existen también otras alternativas, como desplegar directamente sobre el cliente mediante TensorFlow.js o ONNX Runtime Web para Pytorch, o las versiones **Lite/Mobile** de ambos frameworks para desplegarlos en teléfonos móviles. La parte interfaz se puede desarrollar perfectamente en **Flutter o Unity**, según la funcionalidad que le queramos dar, pues ambos entornos son accesibles para Data Scientists (al menos de cara a prototipado) y son compatibles tanto con sistemas operativos Android o MacOS.

### 13.2.3.- Despliegue mediante herramientas cloud

Existen ventajas adicionales si el despliegue de los modelos se lleva a cabo a través de proveedores cloud como AWS, GCP o Azure, pues incorporan diferentes componentes nativos para llevar a cabo estas tareas de forma sencilla e integrable con el resto de la infrastructura. A parte de estos componentes nativos, también son altamente compatibles para incorporar herramientas Python para el despliegue que ya vimos anteriormente (Horovod, Ray, TensorFlowOnSpark, DeepSpeed, BigDL, etc.), así como el pipeline completo de procesamiento de datos, monitorización, validación, etc. y su integración continua a través de DevOps.

### 13.2.4.- Despliegue en sistemas IoT

Cuando queremos desarrollar un modelo cuyos datos (imágenes, sonidos, temperaturas, etc.) se recojan a través de sensores conectados a pequeñas unidades de computación (Rasperry Pi, Jetson Nano, Arduino, etc.) debemos valorar si nos conviene que el modelo se ejecute en dichas unidades o si es mejor que envíen sus datos a la nube y los modelos se ejecuten allí. Entre los factores para decantarnos entre una u otra opción destacan los siguientes:
* Capacidad de procesamiento de la pequeña unidad de computación.
* Velocidad de ejecución, pues desde el dispositivo nos evitamos el paso de enviar y recibir datos.
* Consumo y mantenimiento de la pequeña unidad de computación.
* Escalado, actualizaciones y resolución de incidencias en caso de procesar los modelos en la pequeña unidad de computación.
* Cuestiones de almacenamiento y privacidad del dato, incluyendo la seguridad física.
* Ancho de banda, cobertura y consumo de datos en caso de enviarlos al cloud. El despliegue en sistemas embebidos nos permite no requerir cobertura de internet más que en algunas ocasiones para su mantenimiento, no constantemente.
* Costes, en función a la utilización periódica de los mismos, considerando el coste de mantenimiento y reposición.

En caso de que se prefiera que los modelos se desplieguen en el sistema embebido, existen versiones de TensorFlow y PyTorch para ello, pero en cualquier caso conviene llevar a cabo tareas de cuatización del modelo (simplificación del tipo de datos, reduciendo la precisión), el podado, técnicas de destilación y otras estrategias para reducir su peso computacional, uso de memoria y de disco, etc. 

### 13.2.5.- Revisión de modelos

Una vez que ya hayan llevar a cabo los despliegues, es preciso hacer mediciones periódicas del funcionamiento de los modelos por si ocurriera alguna eventualidad en los datos de entrada, en los modelos o simplemente emepeciera a aparecer **drift** o desgaste en las predicciones. En el caso de disponer de varias versiones de los modelos, también se puede hacer una pos-selección basada en **champion/challengers**, **tests A/B**, algoritmos de **bandido multibrazo** u **optimización bayesiana** para elegir el modelo que mejor esté funcionando. Normalmente se dejan el mejor de ellos en producción y, con fácil acceso, los demás sin ejecución o, si es posible, en **modo sombra** para validar su funcionamiento sin utilizar sus salidas o estimaciones. 

### 13.2.6.- Reutilización de modelos

En ocasiones un mismo modelo puede ser utilizado para varias tareas y por diferentes usuarios, que a veces utilizan frameworks de desarrollo diferentes, por lo que cabe preguntarse cómo gestionar los modelos para su reutilización. Aparte de disponer de una documentación sólida y que permita posteriores evoluciones o variaciones sobre el mismo, se puede paquetizar para el uso compartido de múltiples maneras:
- Mediante el encapsulado en un **microservicio** desplegado en una dirección web, acompañado por su documentación en Swagger.
- Mediante el encapsulado en un **contenedor**, incluyendo todas las dependencias y procedimientos de despliegue dentro del dockerfile.
- Mediante la conversión a un formato de modelo reutilizable por diferentes frameworks, especialmente **ONNX**.
- Mediante herramientas de tracking y gestión de modelos como **MLFlow**, **KubeFlow** o **Comet**.
