## 12.1.- Optimización de hiper-parámetros

### 12.1.1.- Herramientas de código abierto

El aprendizaje automático automatizado (AutoML) es el proceso de automatización de las tareas de aplicación del aprendizaje automático a problemas del mundo real. AutoML incluye potencialmente todas las etapas, desde el comienzo con un conjunto de datos sin procesar hasta la creación de un modelo de aprendizaje automático listo para su implementación. AutoML se propuso como una solución basada en inteligencia artificial para el creciente desafío de aplicar el aprendizaje automático.

El alto grado de automatización en AutoML tiene como objetivo permitir que los no expertos utilicen modelos y técnicas de aprendizaje automático sin necesidad de que se conviertan en expertos en aprendizaje automático. Automatizar el proceso de aplicar el aprendizaje automático de principio a fin ofrece además las ventajas de producir soluciones más simples, una creación más rápida de esas soluciones y modelos que a menudo superan a los modelos diseñados a mano. Las técnicas comunes utilizadas en AutoML incluyen la optimización de hiperparámetros, el meta-aprendizaje y la búsqueda de arquitectura neuronal.

Las principales herramientas de AutoML son:
- **AutoKeras**: sistema AutoML basado en Keras y desarrollado por DATA Lab en Texas A&M Universitycon el fin de que el aprendizaje automático sea accesible para todo el mundo. 
- **AutoGluon**: fácil de usar y de ampliar con un enfoque en el ensamblaje de pilas automatizado, permitiendo realizar rápidamente prototipos de soluciones clásicas y de aprendizaje profundo para sus datos sin procesar con unas pocas líneas de código y sin excesivo conocimiento experto, aprovechando el ajuste automático de hiperparámetros, la selección/ensamblaje de modelos, la búsqueda de arquitectura y el procesamiento de datos.
- **TPOT**: es una herramienta de optimización de canalizaciones basada en árboles que utiliza algoritmos genéticos para optimizar las canalizaciones de aprendizaje automático. TPOT se basa en scikit-learn y utiliza sus propios métodos de clasificación y regresor. TPOT explora miles de canalizaciones posibles y encuentra la que mejor se ajusta a los datos.
- **Optuna**: framework de optimización automática de hiperparámetros, especialmente diseñado para el aprendizaje automático. Cuenta con una API de usuario de estilo definido por ejecución imperativa de alta modularidad en la que el usuario puede construir dinámicamente los espacios de búsqueda de los hiperparámetros.
- **Hyperopt**: biblioteca de Python para la optimización en serie y en paralelo en espacios de búsqueda complicados, que pueden incluir dimensiones de valor real, discretas y condicionales.
- **Neural Newtwork Intelligence (NNI)**: framework de código abierto desarrollado por Microsoft. Se utiliza para automatizar la ingeniería de características, la compresión de modelos, la búsqueda de arquitectura neuronal y el ajuste de hiperparámetros.
- **Auto-Sklearn**: libera a un usuario de aprendizaje automático de la selección de algoritmos y el ajuste de hiperparámetros. Aprovecha las ventajas recientes en la optimización bayesiana, el meta-aprendizaje y la construcción de muestras.

## 12.2.- Aprendizaje con escasez de datos

### 12.2.1.- Transfer learning y fine-tuning

El Transfer Learning es un método de aprendizaje automático en el que un modelo desarrollado para una tarea se reutiliza como punto de partida para un modelo en una segunda tarea. Es un enfoque popular en el aprendizaje profundo en el que se utilizan modelos previamente entrenados como punto de partida en tareas de procesamiento de lenguaje natural y visión por computadora dados los vastos recursos informáticos y de tiempo necesarios para desarrollar modelos de redes neuronales. El Transfer Learning se basa en una idea simple, la de re explotar los conocimientos adquiridos por otras configuraciones (fuentes), para resolver un problema en particular (objetivos). En este contexto, podemos distinguir diferentes abordajes según lo que se quiere transferir, y cuando y como se realiza la transferencia. En términos generales, podemos distinguir 3 tipos de Transfer Learning :
- Aprendizaje por **transferencia inductiva** (Inductive Transfer Learning): En este enfoque, el campo de la fuente y el objetivo es el mismo (mismos datos), pero las tareas de fuente y objetivo son diferentes aunque parecidas. La idea consiste entonces en usar los modelos existentes para reducir de manera ventajosa el campo de aplicación de los modelos posibles (sesgo del modelo), como lo ilustra la figura siguiente:
- Aprendizaje por **transferencia no supervisada** (Unsupervised Transfer Learning): Como en el caso del aprendizaje por transferencia inductiva, los campos de la fuente y el objetivo son similares, aunque las tareas son diferentes. Sin embargo, los datos en ambos campos no están etiquetados. Generalmente es más fácil obtener grandes cantidades de datos no etiquetados, a partir de bases de datos y fuentes en la web por ejemplo, que datos etiquetados. Por ejemplo, el Self-taught clustering es un método que permite realizar el clustering de pequeñas colecciones de datos objetivo no etiquetados, con la ayuda de una gran cantidad de datos fuente no etiquetados.
- Aprendizaje por **transferencia transductiva** (Transductive Transfer Learning): En este método, las tareas fuente y objetivo son similares pero sus campos correspondientes son diferentes en términos de datos o de distribuciones de probabilidad marginales. Por ejemplo, los modelos de NLP, como los que se utiliza para el etiquetado morfosintáctico de palabras, Part-Of-Speech Tagger (POS Tagger) en inglés, son generalmente entrenados y testeados con datos de actualidad como los del Wall Street Journal. Pueden ser adaptados a datos extraídos de redes sociales, cuyo contenido es diferente pero se asemeja al de los periódicos.

Otra variante de gran interés en la actualidad con la llegada de los Transformers es el **Fine-Tuning** de grandes modelos neuronales. En este caso, más que un transfer-learning en que tomamos parte del modelo y actualizamos otro, realmente conservamos el modelo en su práctica totalidad (generalmente enormes) y sólo se modifican componentes de la entrada.

### 12.2.2.- Semi-supervised learning 

El coste asociado al proceso de etiquetado puede hacer que un conjunto de entrenamiento totalmente etiquetado sea inviable, mientras que la adquisición de datos sin etiquetar es relativamente poco costoso. En estos casos, el aprendizaje semi-supervisado puede ser muy útil. El aprendizaje semi-supervisado es una clase de técnicas de aprendizaje automático que utiliza datos de entrenamiento tanto etiquetados como no etiquetados: normalmente una pequeña cantidad de datos etiquetados junto a una gran cantidad de datos no etiquetados. El aprendizaje semi-supervisado se encuentra entre el aprendizaje no supervisado (sin datos de entrenamiento etiquetados) y el aprendizaje supervisado (con todos los datos de entrenamiento etiquetados). Los investigadores del campo del aprendizaje automático han descubierto que los datos no etiquetados, cuando se utilizan junto a una pequeña cantidad de datos etiquetados, pueden mejorar de forma considerable la exactitud del aprendizaje. La adquisición de datos etiquetados para resolver un problema suele requerir un agente humano capacitado para clasificar de forma manual los ejemplos de entrenamiento. 

Un ejemplo de técnica que utiliza aprendizaje semi-supervisado es el **co-entrenamiento**, donde se entrenan dos o más sistemas cada uno en un conjunto de ejemplos, pero de forma que cada sistema utiliza un conjunto de características diferentes (e idealmente independientes) para cada ejemplo. También se puede modelar la distribución de la probabilidad conjunta de las características y las etiquetas. Los datos no etiquetados pueden ser tratados como 'datos que faltan'. Se utiliza de forma común en el algoritmo EM para maximizar la similitud del modelo.

Otra alternativa es la **weakly supervision** o **few shot-learning**, una rama del aprendizaje automático en la que se utilizan fuentes ruidosas, limitadas o imprecisas para proporcionar una señal de supervisión para etiquetar grandes cantidades de datos de entrenamiento en un entorno de aprendizaje supervisado. Este enfoque alivia la carga de obtener conjuntos de datos etiquetados a mano, lo que puede ser costoso o poco práctico. En su lugar, se emplean etiquetas débiles económicas con el entendimiento de que son imperfectas, pero que, sin embargo, pueden usarse para crear un modelo predictivo fuerte. Una herramienta muy utilizada para llevar a cabo esta técnica es Snorkel.

### 12.2.3.- Meta-learning

Meta Learning pregunta: en lugar de comenzar desde cero en cada nueva tarea, ¿hay alguna manera de entrenar un modelo a través de tareas para que la adquisición de nuevas tareas específicas sea más rápida y más eficiente con los datos? Los enfoques en el metaaprendizaje y la disciplina relacionada del aprendizaje de pocos disparos han tomado muchas formas, desde espacios de incrustación agnósticos de tareas de aprendizaje hasta redes recurrentes que se pasan secuencialmente en los datos de entrenamiento y codifican un algoritmo de aprendizaje en sus pesos de evolución de estado, pero posiblemente el más intuitivo de estos métodos ha sido **Model Agnostic Meta Learning (MAML)**. La premisa de MAML es: si queremos un modelo inicial que pueda adaptarse a nuevas tareas extraídas de alguna distribución de tareas, dada una pequeña cantidad de puntos de datos en cada nueva tarea, debemos estructurar nuestro modelo para optimizarlo directamente para ese objetivo.

Existen otras técnicas alternativas y librerías Python de meta-learning, entre las que se encuentran:
- **ADversarial Meta-Learner (ADML)**: aprovecha muestras limpias y antagónicas para optimizar la inicialización de un modelo de aprendizaje de forma antagónica. ADML conduce a las siguientes propiedades deseables: 1) resulta muy efectivo incluso en los casos con solo muestras limpias; 2) es robusto frente a muestras antagónicas, es decir, a diferencia de otros algoritmos de metaaprendizaje, solo conduce a una degradación menor del rendimiento cuando hay muestras antagónicas; 3) arroja luz sobre el abordaje de los casos con muestras limitadas e incluso contaminadas.
- **Context Adaptation for Meta-Learning (CAML)**: método de metaaprendizaje para una adaptación rápida que divide los parámetros del modelo en dos partes, donde los parámetros de contexto que sirven como entrada adicional al modelo y se adaptan en tareas individuales, al tiempo que otros parámetros compartidos que se metaentrenan y se comparten entre tareas.
- **MetaSGD**: similar a SGD y fácilmente entrenable que puede inicializar y adaptar cualquier aprendiz diferenciable en un solo paso, tanto en el aprendizaje supervisado como en el aprendizaje por refuerzo. Meta-SGD tiene una gran capacidad de aprender a aprender, mostrando un rendimiento altamente competitivo.
- **Reptile**: OpenAI desarrolló este sistema de metw-learning, que consiste en muestrear repetidamente una tarea, realizando un descenso de gradiente estocástico en ella y actualizando los parámetros iniciales hacia los parámetros finales aprendidos en esa tarea. Reptile es la aplicación del algoritmo Shortest Descent a la configuración de metaaprendizaje, y es matemáticamente similar al MAML de primer orden.

### 12.2.4.- Active learning

Hay situaciones en las que abundan los datos sin etiquetar, pero el etiquetado manual es costoso. En tal escenario, los algoritmos de aprendizaje pueden consultar activamente al usuario/profesor en busca de etiquetas. Este tipo de aprendizaje supervisado iterativo se denomina aprendizaje activo. Dado que el alumno elige los ejemplos, la cantidad de ejemplos para aprender un concepto a menudo puede ser mucho menor que la cantidad requerida en el aprendizaje supervisado normal. Con este enfoque, existe el riesgo de que el algoritmo se vea abrumado por ejemplos poco informativos. Los desarrollos recientes están dedicados al aprendizaje activo de múltiples etiquetas, el aprendizaje activo híbrido y el aprendizaje activo en un contexto de un solo paso (en línea), combinando conceptos del campo del aprendizaje automático (por ejemplo, conflicto y ignorancia) con políticas de aprendizaje adaptativas e incrementales en el campo del aprendizaje automático en línea.

El Active Learning es un caso especial de aprendizaje automático en el que un algoritmo de aprendizaje puede consultar de forma interactiva a un usuario (o alguna otra fuente de información) para etiquetar nuevos puntos de datos con los resultados deseados. En la literatura estadística, a veces también se le llama diseño experimental óptimo. La fuente de información también se llama maestro u oráculo.

Las dos métricas que dirigen la priorización de etiquetado en el Active Learning son:
- **Incertidumbre**: se trata de priorizar en aquellas etiquetas para las cuales nuestro modelo entrenado con los datos de muestra existentes muestra una mayor incertidumbre en la predicción. De este modo, el etiquetado humano se focalizará en aquellos casos en que el modelo más lo necesita, evitando etiquetar elementos sobre los que el modelo tiene mayor información y capacidad de acierto.
- **Diversidad**: se trata de etiquetar elementos muy distintos entre sí, de modo que etiquetando pocos elementos muy característicos podamos inferir una mayor cantidad de etiquetas y que-estas, además, dispongan de calidad para el entrenamiento (casos positivos y negativos muy distintos entre sí).

Hoy en día, el Active Learning ha mostrado bastantes éxitos como fusión del trabajo humano y de las máquinas, siendo su mejor credencial el **Aprendizaje por Refuerzo a partir de Feed-back Humano (RLHF)**, técnica fundamental para el desarrollo de ChatGPT. 

## 12.3.- Privacidad

### 12.3.1.- Aprendizaje federado

El aprendizaje federado (federated learning) es una técnica de aprendizaje automático que entrena un algoritmo a través de una arquitectura descentralizada formada por múltiples dispositivos los cuales contienen sus propios datos locales y privados. Creado por Google en el año 2017, este enfoque contrasta con las técnicas tradicionales de aprendizaje automático centralizado y de enfoques descentralizados más clásicos, en las que todos los conjuntos de datos se cargan en un servidor de manera idéntica a como están almacenados en local. Gracias a esto se conserva la integridad de la información que está siendo utilizada para el aprendizaje sin poner en peligro la privacidad y seguridad.

El aprendizaje federado es una herramienta muy potente en el sector médico, defensa, telecomunicaciones, IoT o farmacia entre muchos otros campos, pues el principio general de este aprendizaje consiste en entrenar modelos locales sobre muestras de datos también locales e intercambiar con cierta frecuencia parámetros (por ejemplo, los pesos de una red neuronal profunda) entre estos nodos locales para generar un modelo global compartido por todos los nodos.

La principal diferencia entre el aprendizaje federado y el distribuido radica en los supuestos que se hacen sobre las propiedades de los conjuntos de datos locales ya que el aprendizaje distribuido tiene como objetivo original paralelizar la potencia de cálculo, mientras que el aprendizaje federado tiene como objetivo original el **entrenamiento en conjuntos de datos heterogéneos y de forma privada**. Los clientes que participan en el aprendizaje federado pueden ser poco fiables, ya que están sujetos a más fallos o abandonos, ya que suelen depender de medios de comunicación menos potentes y de sistemas alimentados por baterías. En contraparte, en el aprendizaje distribuido los nodos suelen ser centros de datos que tienen potentes capacidades de cálculo y están conectados entre sí con redes rápidas.

Las herramientas más habituales para desarrollar sistemas de aprendizaje federado son:
- **PySyft**: desarrollada sobre PyTorch por Openmined e incluyendo el framework más completo para garantizar la seguridad.
- **TensorFlow Federated (TFF)**: es un marco de código abierto para el aprendizaje automático y otros cálculos en datos descentralizados. Por ejemplo, TFF se ha utilizado para entrenar modelos de predicción para teclados móviles sin cargar datos confidenciales de escritura en los servidores.
- **Flower**: librería de python que proporciona interfaces para que los desarrolladores implementen su algoritmo de entrenamiento tanto en PyTorch, TensorFlow como en MXNET.

El aprendizaje federado suele diferenciarse en dos tipos:
* Aprendizaje federado **horizontal**: consiste en compartir registros procedentes de diferentes fuentes y aprender de forma distribuida.
* Aprendizaje federado **vertical**: consiste en compartir columnas procedentes de diferentes fuentes y aprender de forma distribuida.

### 12.3.2.- Mecanismos de privacidad

Existen varias técnicas para preservar la privacidad del dato. Los mecanismos de privacidad más extendidos son:

- **Secure multiparty computation (SMPC)**: aborda las preocupaciones de varias organizaciones que desean ejecutar algún cálculo que involucre todos sus datos sin confiar entre sí o en un servidor central para (1) almacenar los datos o (2) realizar el cálculo. La suposición es que los datos de entrada son confidenciales, pero la salida (y cualquier información que implique sobre la entrada) no es confidencial.
- **Fully Homomorphic Encryption (FHE)**: En el cifrado tradicional, los datos normalmente se cifran en reposo y en movimiento. Sin embargo, los datos deben descifrarse para que se ejecute un cálculo. El cifrado totalmente homomórfico (FHE) es una técnica que permite cálculos arbitrarios en un conjunto de datos cifrados sin necesidad de descifrar el conjunto de datos, incluso al ejecutar el cálculo. FHE impone un paradigma en el que el analista o el usuario final tiene acceso total y sin restricciones a los datos subyacentes, pero el entorno informático/de almacenamiento nunca tiene acceso a los datos en claro (porque siempre están encriptados). FHE es útil en situaciones en las que una organización quiere usar el hardware físico de otra organización (como un proveedor de computación en la nube) pero cree que el proveedor de la nube corre el riesgo de verse comprometido. 
- **Differential privacy (DP)**: permite estadísticas y aprendizaje automático en un conjunto de datos al tiempo que garantiza que la información sobre registros individuales en el conjunto de datos no se pueda extraer ni inferir. Supongamos que una empresa tiene información confidencial sobre sus clientes y desea que un tercero pueda desarrollar informes y modelos predictivos sobre los datos. Para preservar el valor de los datos, la empresa quiere asegurarse de que toda la información sobre sus clientes se pueda utilizar en los informes y modelos. Sin embargo, a la empresa le preocupa que el tercero pueda obtener acceso a información confidencial sobre sus clientes mientras consulta y modela los datos.

### 12.4.2.- Responsible AI

Con el uso creciente de los sistemas de IA en todos los dominios, surgen muchas preguntas sobre la ética, la confianza, la legalidad y el gobierno de datos de la IA. Las decisiones impulsadas por la IA deben evaluarse en diferentes frentes, como los riesgos comerciales, la privacidad de los datos, los problemas de salud y seguridad y la igualdad.

El área de Responsible AI (RAI) consiste en diseñar prácticas para desarrollar, implementar y escalar la IA para causas constructivas y buenas intenciones para impactar positivamente a las personas y la sociedad. Fomenta la confianza de las personas en el sistema de IA. De este modo se pretende ayudar a transformar las aplicaciones de IA en más responsables, éticas y transparentes. Evalúa los esfuerzos organizacionales de IA desde el punto de vista ético y legal.

Las aplicaciones de IA son críticas para el negocio y se ocupan de datos confidenciales. Por lo tanto, las personas deben comprender en profundidad el papel de la IA. Surgien preguntas como "¿La IA está apoyando decisiones precisas?", "¿Está violando alguna política o privacidad?", "¿Existe algún mecanismo para monitorear esta tecnología y sus resultados?". Estas preguntas se abordan de manera efectiva mediante un nuevo marco de gobernanza: IA responsable. Con la IA responsable, las empresas establecen objetivos clave y establecen estrategias de gobierno para las iniciativas de IA. RAI permite:

* **Sesgo reducido** en conjuntos de datos: ayuda a garantizar que tanto el algoritmo como los datos subyacentes sean imparciales y representativos de la realidad.
* **IA ética**: brinda un enfoque que prioriza la seguridad para garantizar el uso ético de los datos. Protege la privacidad y seguridad de sus datos confidenciales para evitar su uso no ético por cualquier medio. Ayuda a mitigar el riesgo y beneficia a las personas, las organizaciones y la sociedad en su conjunto.
* **Transparencia**: impulsa la transparencia en todos los procesos y funciones. Permite explicaciones comprensibles para los humanos de las predicciones hechas en contraste con el aprendizaje automático tradicional de caja negra.
* **Gobernanza efectiva**: los procesos de desarrollo de ML deben documentarse para evitar su alteración por mala intención.
* **Adaptabilidad**: los modelos que respaldan las iniciativas de IA deben adaptarse a entornos complejos sin introducir sesgos.

## 12.5.- Aprendizaje profundo probabilístico

### 12.5.1.- Tensorflow Probability

Hasta ahora, hemos visto que la mayor parte de los modelos neuronales ofrecen estimaciones escalares, a excepción de las capas de salida softmax para categorías, pero en ocasiones podríamos necesitar que el modelo ofreciera una estimación del margen de confianza de dicha predicción en forma de función de probabilidad.

TensorFlow Probability (TFP) es una biblioteca de Python basada en TensorFlow que facilita la combinación de modelos probabilísticos y aprendizaje profundo en hardware moderno (TPU, GPU). Es para científicos de datos, estadísticos, investigadores de ML y profesionales que desean codificar el conocimiento del dominio para comprender los datos y hacer predicciones. LTFP incluye:
* Una amplia selección de distribuciones de probabilidad y biyectores.
* Herramientas para construir modelos probabilísticos profundos, incluidas capas probabilísticas y la probabilidad conjunta.
* Inferencia variacional y cadena de Markov Monte Carlo.
* Optimizadores como Nelder-Mead, BFGS y SGLD.

Dado que TFP hereda los beneficios de TensorFlow, puede crear, adaptar e implementar un modelo utilizando un solo idioma durante todo el ciclo de vida de la exploración y producción del modelo.

```python
import tensorflow as tf
import tensorflow_probability as tfp

# Pretend to load synthetic data set.
features = tfp.distributions.Normal(loc=0., scale=1.).sample(int(100e3))
labels = tfp.distributions.Bernoulli(logits=1.618 * features).sample()

# Specify model.
model = tfp.glm.Bernoulli()

# Fit model given data.
coeffs, linear_response, is_converged, num_iter = tfp.glm.fit(
    model_matrix=features[:, tf.newaxis],
    response=tf.cast(labels, dtype=tf.float32),
    model=model)
```

### 12.5.2.- Bayesian Deep Learning

La propiedad distintiva clave de un enfoque bayesiano es la marginación, en lugar de que usar un solo ajuste de pesos. La marginación bayesiana puede particularmente mejorar la precisión y la calibración de las modernas redes neuronales profundas, que son típicamente subespecificado por los datos, y puede representar muchas soluciones convincentes pero diferentes. El enfoque bayesiano propone un enfoque relacionado que mejora la distribución predictiva al marginar dentro de las cuencas de atracción, sin gastos generales significativos.

Para ello se investigan las funciones previas implícitas por una vaga distribución sobre los pesos de las redes neuronales, lo que explica la generalización propiedades de tales modelos desde una perspectiva probabilística. Desde esta perspectiva, se explican los resultados que se han presentado como misteriosos y distintos a los neurales generalización de redes, como la capacidad de ajustar imágenes con etiquetas aleatorias, y muestran que estos resultados se pueden reproducir con procesos gaussianos. también mostramos que el promedio del modelo bayesiano alivia el doble descenso, lo que resulta en monótona
mejoras de rendimiento con mayor flexibilidad.

### 12.6.2.- Aprendizaje profundo con Quantum Computing

El aprendizaje automático cuántico (**Quantum machine learning**) suma esfuerzos para combinar la mecánica cuántica y el aprendizaje automático. Los algoritmos o modelos de aprendizaje automático cuántico intentan usar las ventajas de la información cuántica con el fin de mejorar el aprendizaje automático clásico, por ejemplo desarrollando implementaciones eficientes de pesados algoritmos clásicos mediante computación cuántica. Además, el aprendizaje automático cuántico incluye también la aproximación recíproca, aplicando los métodos de aprendizaje automático clásico a la teoría de la información cuántica.

Al igual que su homólogo clásico, el aprendizaje automático cuántico incluye técnicas de aprendizaje supervisado cuántico, aprendizaje no supervisado cuántico y aprendizaje por refuerzo cuántico. Es especialmente interesante la aportación que se puede hacer desde el QML para entrenar modelos complejos basados en **aprendizaje por refuerzo**, pues son difíciles de entrenar por métodos tradicionales y la computación cuántica ofrece interesantes alternativas.

Con la llegada de las tecnologías NISQ (Computadores cuánticos ruidosos de escala intermedia) se ha intentado reemplazar las redes neuronales por circuitos cuánticos. En una red neuronal, la función es aproximada ajustando el parámetro del peso en cada capa, y el concepto es exactamente el mismo en QCL. Se proporcionan datos de entrada a un circuito cuántico y se ajusta iterativamente los parámetros del circuito para que el circuito optimizado proporcione la salida deseada.

### 12.6.3.- Aprendizaje profundo con Blockchain

La intersección del aprendizaje profundo con Blockchain es bidireccional:
- Blockchain a Deep Learning: mediante tecnologías de blockchain como OpenMined es posible compartir modelos al tiempo que se preserva la propiedad intelectual de los autores de un modelo.
- Deep Learning a Blockchain: aparte de ayudar a optimizar los tiempos de minado o la gestión de carteras de criptoactivos, el aprendizaje por refuerzo puede ayudar a generar NFTs o generar modelos predictivos que pudiera ser almacenados automáticamete en la cadena de bloques.
