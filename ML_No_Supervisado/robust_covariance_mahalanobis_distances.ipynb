
# Robust covariance estimation and Mahalanobis distances relevance

This example shows covariance estimation with Mahalanobis
distances on Gaussian distributed data.

For Gaussian distributed data, the distance of an observation
$x_i$ to the mode of the distribution can be computed using its
Mahalanobis distance:

\begin{align}d_{(\mu,\Sigma)}(x_i)^2 = (x_i - \mu)^T\Sigma^{-1}(x_i - \mu)\end{align}

where $\mu$ and $\Sigma$ are the location and the covariance of
the underlying Gaussian distributions.

In practice, $\mu$ and $\Sigma$ are replaced by some
estimates. The standard covariance maximum likelihood estimate (MLE) is very
sensitive to the presence of outliers in the data set and therefore,
the downstream Mahalanobis distances also are. It would be better to
use a robust estimator of covariance to guarantee that the estimation is
resistant to "erroneous" observations in the dataset and that the
calculated Mahalanobis distances accurately reflect the true
organization of the observations.

The Minimum Covariance Determinant estimator (MCD) is a robust,
high-breakdown point (i.e. it can be used to estimate the covariance
matrix of highly contaminated datasets, up to
$\frac{n_\text{samples}-n_\text{features}-1}{2}$ outliers)
estimator of covariance. The idea behind the MCD is to find
$\frac{n_\text{samples}+n_\text{features}+1}{2}$
observations whose empirical covariance has the smallest determinant,
yielding a "pure" subset of observations from which to compute
standards estimates of location and covariance. The MCD was introduced by
P.J.Rousseuw in [1]_.

This example illustrates how the Mahalanobis distances are affected by
outlying data. Observations drawn from a contaminating distribution
are not distinguishable from the observations coming from the real,
Gaussian distribution when using standard covariance MLE based Mahalanobis
distances. Using MCD-based
Mahalanobis distances, the two populations become
distinguishable. Associated applications include outlier detection,
observation ranking and clustering.

## Generate data

First, we generate a dataset of 125 samples and 2 features. Both features
are Gaussian distributed with mean of 0 but feature 1 has a standard
deviation equal to 2 and feature 2 has a standard deviation equal to 1. Next,
25 samples are replaced with Gaussian outlier samples where feature 1 has
a standard deviation equal to 1 and feature 2 has a standard deviation equal
to 7.



```python
import numpy as np

# for consistent results
np.random.seed(7)

n_samples = 125
n_outliers = 25
n_features = 2

# generate Gaussian data of shape (125, 2)
gen_cov = np.eye(n_features)# matriz de covarianza identidad de tamaño n_features x n_features
#np.eye crea una matriz identidad de tamaño n_features x n_features
gen_cov[0, 0] = 2.0# matriz de covarianza para los datos normales
X = np.dot(np.random.randn(n_samples, n_features), gen_cov)# datos generados a partir de una distribución normal multivariante con media 0 y covarianza gen_cov
#np.dot realiza el producto punto entre dos matrices, np.random.randn genera una matriz de números aleatorios con distribución normal estándar

# # add some outliers
outliers_cov = np.eye(n_features)# matriz de covarianza identidad de tamaño n_features x n_features
outliers_cov[np.arange(1, n_features), np.arange(1, n_features)] = 7.0# matriz de covarianza para los outliers
#np.arange crea un array con valores desde 1 hasta n_features-1
X[-n_outliers:] = np.dot(np.random.randn(n_outliers, n_features), outliers_cov)
# datos de outliers generados a partir de: una distribución normal multivariante con media 0 y covarianza outliers_cov
```

## Comparison of results

Below, we fit MCD and MLE based covariance estimators to our data and print
the estimated covariance matrices. Note that the estimated variance of
feature 2 is much higher with the MLE based estimator (7.5) than
that of the MCD robust estimator (1.2). This shows that the MCD based
robust estimator is much more resistant to the outlier samples, which were
designed to have a much larger variance in feature 2.



```python
import matplotlib.pyplot as plt

from sklearn.covariance import EmpiricalCovariance, MinCovDet

# fit a MCD robust estimator to data
robust_cov = MinCovDet().fit(X)# ajuste de un estimador robusto MCD a los datos, MCD es Minimum Covariance Determinant
#minCovDet calcula la matriz de covarianza robusta de los datos
# fit a MLE estimator to data
emp_cov = EmpiricalCovariance().fit(X)# ajuste de un estimador MLE a los datos, MLE es Maximum Likelihood Estimator
#empiricalCovariance calcula la matriz de covarianza empírica de los datos
print(
    "Estimated covariance matrix:\nMCD (Robust):\n{}\nMLE:\n{}".format(
        robust_cov.covariance_, emp_cov.covariance_# matriz de covarianza estimada por ambos métodos
    )
)
```

To better visualize the difference, we plot contours of the
Mahalanobis distances calculated by both methods. Notice that the robust
MCD based Mahalanobis distances fit the inlier black points much better,
whereas the MLE based distances are more influenced by the outlier
red points.



```python
import matplotlib.lines as mlines

fig, ax = plt.subplots(figsize=(10, 5))# creación de una figura y un conjunto de ejes para la visualización
# Plot data set
inlier_plot = ax.scatter(X[:, 0], X[:, 1], color="black", label="inliers")# dibujo de los puntos de datos normales
outlier_plot = ax.scatter(# dibujo de los puntos de datos atípicos
    X[:, 0][-n_outliers:], X[:, 1][-n_outliers:], color="red", label="outliers"
)
ax.set_xlim(ax.get_xlim()[0], 10.0)# ajuste de los límites del eje x
ax.set_title("Mahalanobis distances of a contaminated data set")

# Create meshgrid of feature 1 and feature 2 values
xx, yy = np.meshgrid(#meshgrid crea una cuadrícula a partir de dos vectores de coordenadas
    np.linspace(plt.xlim()[0], plt.xlim()[1], 100),# creación de un vector de 100 puntos entre los límites actuales del eje x
    np.linspace(plt.ylim()[0], plt.ylim()[1], 100),# creación de un vector de 100 puntos entre los límites actuales del eje y
)
zz = np.c_[xx.ravel(), yy.ravel()]#ravel aplana las matrices xx e yy, np.c_ concatena las dos matrices a lo largo de las columnas
# Calculate the MLE based Mahalanobis distances of the meshgrid
mahal_emp_cov = emp_cov.mahalanobis(zz)# cálculo de las distancias de Mahalanobis basadas en MLE para los puntos de la cuadrícula
#emp_cov.mahalanobis calcula las distancias de Mahalanobis utilizando la matriz de covarianza empírica
mahal_emp_cov = mahal_emp_cov.reshape(xx.shape)# reshape de las distancias para que coincidan con la forma de la cuadrícula
#mahal_emp_cov.reshape cambia la forma del array de distancias para que coincida con la forma de la cuadrícula
emp_cov_contour = plt.contour(# dibujo de las líneas de contorno para las distancias de Mahalanobis basadas en MLE
    #countour dibuja las líneas de contorno en el gráfico
    xx, yy, np.sqrt(mahal_emp_cov), cmap=plt.cm.PuBu_r, linestyles="dashed"#MLE es Maximum Likelihood Estimator
    #np.sqrt calcula la raíz cuadrada de las distancias de Mahalanobis
    #cmap=plt.cm.PuBu_r establece el mapa de colores para las líneas de contorno, puBu_r es un mapa de colores que va de púrpura a azul claro
    #linestyles="dashed" establece el estilo de línea para las líneas de contorno
)
# Calculate the MCD based Mahalanobis distances
mahal_robust_cov = robust_cov.mahalanobis(zz)# cálculo de las distancias de Mahalanobis basadas en MCD para los puntos de la cuadrícula
#robust_cov.mahalanobis calcula las distancias de Mahalanobis utilizando la matriz de covarianza robusta
#MCD es Minimum Covariance Determinant
mahal_robust_cov = mahal_robust_cov.reshape(xx.shape)# reshape de las distancias para que coincidan con la forma de la cuadrícula
#mahal_robust_cov.reshape cambia la forma del array de distancias para que coincida con la forma de la cuadrícula
robust_contour = ax.contour(# dibujo de las líneas de contorno para las distancias de Mahalanobis basadas en MCD
    xx, yy, np.sqrt(mahal_robust_cov), cmap=plt.cm.YlOrBr_r, linestyles="dotted"
)

# Add legend
ax.legend(
    [#mlines.Line2D crea una línea personalizada para la leyenda
        mlines.Line2D([], [], color="tab:blue", linestyle="dashed"),
        mlines.Line2D([], [], color="tab:orange", linestyle="dotted"),
        inlier_plot,# referencia al gráfico de puntos normales
        outlier_plot,# referencia al gráfico de puntos atípicos
    ],
    ["MLE dist", "MCD dist", "inliers", "outliers"],
    loc="upper right",
    borderaxespad=0,# espacio entre la leyenda y los ejes
)

plt.show()
```

Finally, we highlight the ability of MCD based Mahalanobis distances to
distinguish outliers. We take the cubic root of the Mahalanobis distances,
yielding approximately normal distributions (as suggested by Wilson and
Hilferty [2]_), then plot the values of inlier and outlier samples with
boxplots. The distribution of outlier samples is more separated from the
distribution of inlier samples for robust MCD based Mahalanobis distances.



```python
fig, (ax1, ax2) = plt.subplots(1, 2)
plt.subplots_adjust(wspace=0.6)

# Calculate cubic root of MLE Mahalanobis distances for samples
emp_mahal = emp_cov.mahalanobis(X - np.mean(X, 0)) ** (0.33)# cálculo de las distancias de Mahalanobis basadas en MLE para los puntos de datos
#np.mean(X, 0) calcula la media de los datos a lo largo de las filas
#emp_cov.mahalanobis calcula las distancias de Mahalanobis utilizando la matriz de covarianza empírica
# Plot boxplots
ax1.boxplot([emp_mahal[:-n_outliers], emp_mahal[-n_outliers:]], widths=0.25)# diagrama de caja para las distancias de Mahalanobis basadas en MLE
#[emp_mahal[:-n_outliers] toma las distancias de Mahalanobis de los puntos normales, emp_mahal[-n_outliers:] toma las distancias de los outliers
# Plot individual samples
ax1.plot(
    np.full(n_samples - n_outliers, 1.26),
    emp_mahal[:-n_outliers],
    "+k",
    markeredgewidth=1,# estilo del borde del marcador
)
ax1.plot(np.full(n_outliers, 2.26), emp_mahal[-n_outliers:], "+k", markeredgewidth=1)# estilo del borde del marcador
ax1.axes.set_xticklabels(("inliers", "outliers"), size=15)# etiquetas del eje x
ax1.set_ylabel(r"$\sqrt[3]{\rm{(Mahal. dist.)}}$", size=16)# etiqueta del eje y
ax1.set_title("Using non-robust estimates\n(Maximum Likelihood)")

# Calculate cubic root of MCD Mahalanobis distances for samples
robust_mahal = robust_cov.mahalanobis(X - robust_cov.location_) ** (0.33)# cálculo de las distancias de Mahalanobis basadas en MCD para los puntos de datos
#robust_cov.location_ es la media robusta calculada por el estimador MCD
#robust_cov.mahalanobis calcula las distancias de Mahalanobis utilizando la matriz de covarianza robusta
# Plot boxplots
ax2.boxplot([robust_mahal[:-n_outliers], robust_mahal[-n_outliers:]], widths=0.25)# diagrama de caja para las distancias de Mahalanobis basadas en MCD
# Plot individual samples
ax2.plot(
    np.full(n_samples - n_outliers, 1.26),
    robust_mahal[:-n_outliers],# diagrama de caja para las distancias de Mahalanobis basadas en MCD
    "+k",# estilo del borde del marcador
    markeredgewidth=1,# estilo del borde del marcador
)
ax2.plot(np.full(n_outliers, 2.26), robust_mahal[-n_outliers:], "+k", markeredgewidth=1)# estilo del borde del marcador
ax2.axes.set_xticklabels(("inliers", "outliers"), size=15)# etiquetas del eje x
ax2.set_ylabel(r"$\sqrt[3]{\rm{(Mahal. dist.)}}$", size=16)# etiqueta del eje y
ax2.set_title("Using robust estimates\n(Minimum Covariance Determinant)")

plt.show()
```
