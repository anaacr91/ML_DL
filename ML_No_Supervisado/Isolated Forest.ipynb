```python
print(__doc__)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

rng = np.random.RandomState(42)

# Generate train data
X = 0.3 * rng.randn(100, 2)# datos normales centrados en el origen
X_train = np.r_[X + 2, X - 2]# datos normales desplazados a (2,2) y (-2,-2)
# Generate some regular novel observations
X = 0.3 * rng.randn(20, 2)# datos normales centrados en el origen
X_test = np.r_[X + 2, X - 2]# datos normales desplazados a (2,2) y (-2,-2)
# Generate some abnormal novel observations
X_outliers = rng.uniform(low=-4, high=4, size=(20, 2))# datos uniformes en el rango [-4,4] en ambas dimensiones

# fit the model
clf = IsolationForest(max_samples=100, random_state=rng)# crear el modelo de IsolationForest
clf.fit(X_train)# ajustar el modelo a los datos de entrenamiento
y_pred_train = clf.predict(X_train)# predecir los datos de entrenamiento
y_pred_test = clf.predict(X_test)# predecir los datos de prueba
y_pred_outliers = clf.predict(X_outliers)# predecir los datos de outliers

# plot the line, the samples, and the nearest vectors to the plane
xx, yy = np.meshgrid(np.linspace(-5, 5, 50), np.linspace(-5, 5, 50))# crear una malla de coordenadas para el gráfico de contorno
Z = clf.decision_function(np.c_[xx.ravel(), yy.ravel()])# puntuaciones de decision del modelo para cada punto de la malla
#desicion_function devuelve la puntuación de decisión del modelo para cada muestra en la malla
#ravel aplana las matrices xx e yy en vectores
#np.c_ concatena los vectores a lo largo de la segunda dimensión para crear
Z = Z.reshape(xx.shape)

plt.title("IsolationForest")
plt.contourf(xx, yy, Z, cmap=plt.cm.Blues_r)

b1 = plt.scatter(X_train[:, 0], X_train[:, 1], c='white',
                 s=20, edgecolor='k')
b2 = plt.scatter(X_test[:, 0], X_test[:, 1], c='green',
                 s=20, edgecolor='k')
c = plt.scatter(X_outliers[:, 0], X_outliers[:, 1], c='red',
                s=20, edgecolor='k')
plt.axis('tight')
plt.xlim((-5, 5))
plt.ylim((-5, 5))
plt.legend([b1, b2, c],
           ["training observations",
            "new regular observations", "new abnormal observations"],
           loc="upper left")
plt.show()
```
