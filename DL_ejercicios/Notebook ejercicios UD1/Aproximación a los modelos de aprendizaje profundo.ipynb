## 2.1.- Glosario de modelos neuronales

### 2.1.1.- Mapa general de modelos

Los principales modelos de redes neuronales profundas existentes en la actualidad son lo siguientes:

* **Modelos fully-connected / dense (FCNN)**: extensión del número de capas en los modelos MLP.
* **Modelos convolucionales (CNN)**: modelos basados en convoluciones empleados principalmente en visión por computador.
* **Modelos recurrentes (RNN)**: modelos neuronales con memoria empleados principalmente para la predicción de secuencias.
* **Autoencoders profundos (DAE)**: modelización de las propias entradas con el fin de reducir la dimensionalidad.
* **Modelos de redes generativas antagónicas (GAN)**: generación de imágenes o textos mediante redes compitiendo entre sí.
* **Modelos profundos de aprendizaje por refuerzo (DeepRL)**: modelos para optimizar la interacción de con nuestro entorno. 
* **Modelos de redes siamesas (SiamNN)**: modelos para comparar pares de entradas y obtener su semejanza mutua. 
* **Modelos de redes neuronales profundas en grafos (GNN)**: modelos para tratar entradas con estructura de grafo.
* **Modelos de Transformers**: modelos avanzados para generar textos o imágenes, así como para memorizar dominios.

### 2.1.2.- Modelos fully-connected / dense (FCNN)

Se trata de la versión profunda de los modelos MLP (perceptrón multi-capa), los cuales se pueden utilizar para modelos supervisados de regresión y clasificación en fenómenos no lineales. Lo más habitual, sin embargo, es combinar únicamente capas de este tipo de modelos con otros capas de otro tipo. Se emplean también a menudo en la última capa (head), lo que permite entrenar modelos a partir de backbones de redes profundas de otro tipo como embedding y la capa fully-conneted como parte supervisada para las últimas capas.

<img src="images/1_PRS1mHOEHdeRJajZ6H7ebg.png" width="600" height="400" style="float:center" />

### 2.1.3.- Modelos convolucionales (Convolutional Neural Networks, CNN)

Se trata de un tipo de red neuronal artificial en la que los pesos no son precisamente esclares, sino capas complejas basadas en kernels matriciales que convolucionan con los valores de las neuronas existentes en la capa anterior o con agregaciones espaciales de las mismas, a imagen y semejanza a como operan las neuronas en la corteza visual de un cerebro biológico. Este tipo de redes son muy efectivas para tareas de visión artificial, como en la clasificación y segmentación de imágenes, entre otras aplicaciones.

<img src="images/Typical_cnn.png" width="500" height="400" />

### 2.1.4.- Modelos recurrentes (Recurrent Neural Networks, RNN)

Se trata de un tipo de red neuronal y de capa en el que las neuronas no sólo se conectan unidireccionalmente hacia la salida, sino que también lo hacen de forma hacia los lados (neuronas de la predicción anterior en una secuencia), a modo de memoria tal y como inspiraron los modelos de Hopfield. Se trata de modelos muy empleados para la modelización de secuencias, desde series temporales hasta audios como secuencias de vibraciones y textos como secuencias de palabras. Los principales modelos recurrentes en la actualidad son los LSTM (Long Short-term Memory) y los GRU (Gated Recurrent Units), pudiendo tratar secuencias muy largas, elemento a elemento ya que recuerdan las salidas anteriores como entrada para la siguiente predicción.

<img src="images/rnn.png" width="500" height="400" />

### 2.1.5.- Autoencoders profundos (Deep Autoencoder, DAE)

Se trata de un tipo de red neuronal de tipo no supervisado utilizada para aprender codificaciones de datos de entrada, de modo que la entrada de estos modelos coincide con su salida. Es decir, este tipo de modelos modelo trata de modelizar sus propias entradas y, aunque pudiera parecer absurdo a primera vista, este tiene múltiples funcionalidades, desde reducir la dimensión de dichas entradas de forma no lineal a la reducción de ruido y la generación de datos, entre otros. Un tipo muy especial de autoencoder es el Autoencoder Variacional, también empleado dentro del campo de la computación cuántica, el cual se emplea a menudo para la generación de imágenes, texto u otras representaciones de datos complejas.

<img src="images/Autoencoder_structure.png" width="500" height="400" />

### 2.1.6.- Modelos de red generativa antagónica (Generative Adversarial Network, GAN)

Se trata de un sistema de dos redes neuronales compitiendo entre sí, el cual se utiliza para generar datos complejos, tales como textos o imágenes. Las dos redes neuronales son la parte generativa, que inventa candidatos a imágenes o textos, y la discriminativa, que aprende a distinguir si la imagen es inventada o real. Haciendo competir iterativamente a una con la otra, la parte generador va a prendiendo por ensayo y error a generar imágenes o textos cada vez más difíciles de distinguir por la red generativa, llegándose a crear imágenes y textos muy complejos y aparentemente reales tanto para la red discriminativa como para el ojo humano.

### 2.1.7.- Modelos profundos de aprendizaje por refuerzo (Deep Reinforcement Learning, DRL)

Se trata de modelos utilizados para entrenar a sistemas inteligentes a interactuar de forma óptima dentro de un entorno, como por ejemplo un vehículo autónomo dentro de la carretera, un chatbot dentro de una conversación o un bot de bolsa dentro de mercados financieros en tiempo real. Estos modelos se caracterizan por ser la versión profunda de los modelos clásicos de aprendizaje por refuerzo (Deep Reinforcement Learning), especialmente q-learning, policy optimization o los métodos actor-critic.

<img src="images/deeprl.png" width="500" height="400" />

### 2.1.8.- Redes siamesas (SiamNN)

Se trata de una estructura de red neuronal que sirve para estimar si dos objetos pertenecen a la misma clase. Las redes siamesas están compuesta por una capa de entrada en la que se introducen dos imágenes o dos textos, una serie de capas intermedias con pesos compartidos que determinan un embedding para cada miembro del par y una función de distancia como capa de salida indicando si dichas entradas pertenecen o no a la misma clase.

<img src="images/siamese.png" width="500" height="400" />

### 2.1.9.- Modelos neuronales de grafos (Graph neural networks, GNN)

Se trata de modelos de redes neuronales en grafos (Graph Neural Networks), en fuerte etapa de desarrollo en la actualidad, que destacan especialmente porque permiten trascender la dimensión geométrica espacial de las redes convolucionales y atender a todo tipo de relaciones locales en de neuronas dentro de un grafo, pudiendo ser utilizadas, entre otras cosas, para clasificación de objetos en imágenes 3D, la clasificación de esrtucturas complejas de moléculas o compuestos químicos o para predecir la existencia o no de nuevas aristas o nodos dentro de grafos, dotando a los modelos de redes neuronaes profundas de una mayor flexibilidad a la hora de disponer las entradas y, sobre todo, las interconexiones entre ellas.

<img src="images/GNN_building_blocks.png" width="500" height="400" />

### 2.1.10.- Transformers

Se trata de los modelos que mayor éxito están cosechando en la actualidad, casi reemplazando cada vez más tanto a las redes recurrentes como a las convolucionales. La innovación de estos modelos consistió en sacar provecho de las capas de "atención", las cuales permiten analizar secuencias y contextos de entradas de forma no necesariamente secuencial, sino estableciendo pesos entre ellas en forma de tensores. Se trata de modelos que en la actualidad se están entrenando por medio de uso de sistemas informáticos extremadamente potentes, ya que contienen centeneres de millones de neuronas interconectadas y volúmenes de datos asimismo inmensos. Son la base de modelos actuales como para la representación de texto (GPT-3 o ChatGPT) o de imágenes (Stable Diffusion o DALL-e), los cuales no se pueden entrenar en máquinas de escritorio ni portátiles, pero sí se las puede realizar fine-tuning para adaptarlas a diferentes dominios más específicos a determinar por el usuario.

## 2.2.- Modelos neuronales fully-connected (MLP)

### 2.2.1.- Pesos y funciones de activación

En la clase de hoy vamos a analizar y practicar con la arquitectura de modelo neuronal más sencila: la Fully-Connected Neural Network (FCNN). Los elementos principales de este tipo de redes son:
- <ins>Entradas (i) y salidas (o)</ins>, generalmente valores numéricos escalares o dummies de variables categóricas.
- <ins>Pesos (w)</ins>, uno por cada conexión entre neuronas. También puede haber conexiones con un término constante exterior, llamado "bias" (sesgo). Las conexiones suelen producirse entre capas consecutivas, pero en ocasiones se pueden conectar saltando alguna capa o directamente con la salida (conexiones directas).
- <ins>Funciones de activación</ins>, la cual suele ser una función creciente sobre la suma de las entradas enteriores multiplicadas cada una por su peso. Las funciones de activación más tradicionales son de tipo escalón (las más antiguas) y, para que sea derivable, sigmoide, tanto softmax como tangente hiperbólica. En la actualidad, las funciones de acivación que mejores resultados dan son las de ReLU (Rectified Linear Unit) y Leakly ReLU.
- <ins>Ratio de aprendizaje</ins>: indica la intensidad con la que se van a actualizar los pesos del modelo, de modo que, si es su valor demasiado bajo tardará mucho en aprender, mientras que, si es demasiado alto, empezará a oscilar sin llegar a aprender. Para evitar este tipo de situaciones, con el tiempo se han ido desarrollando diferentes algoritmos de optimización como el AdaGrad, AdaDelta, AdaM o los momentos de Nesterov.
- <ins>Lotes o batches</ins>: tamaño del conjunto de datos sobre los que se medirá el error y se actualizarán los pesos, siendo conveniente en general que no sea uno, pues no daría tiempo a aprender de forma sólida, o muy alto, pues no llegaría a distinguir los valores de los pesos con el detalle suficiente. 
- <ins>Transformaciones sobre los datos de entrada (features)</ins>: no suele ser necesario en aprendizaje profundo, pero en este tipo de modelos se pueden llegar a utilizar, tal y como veremos en las prácticas que haremos con la aplicación tensorflow/playground.

### 2.2.2.- Algoritmo de entrenamiento

La estrategia de entrenar la red neuronal es en principio similar a la de otros modelos de machine learning conocidos. En primer lugar necesitamos definir una **función de pérdida** que nos indique cómo de lejos nuestra predicción está del valor real, de modo que necesitaremos un algoritmo para minimizar este error.
Al igual que ocurre en otros modelos como la regresión lineal, podríamos intentar hacer la derivada en función a cada parámetro y tratar de encontrar para qué valores estas derivadas son cero. Tal y como sucede en funciones de múltiples variables con no linearidades, este sistema de ecuaciones es difícil de resolver, por lo que se opta por métodos iterativos relacionados con el **descenso por el gradiente**. Como se trata de encontrar el mínimo de la función de pérdidas, debemos movernos en dirección contraria al gradiente, tal y como se ve en la figura de la izquierda.

Al momento de computar todas las derivadas parciales de la pérdida en función a cada parámetro (peso) de la red en las distintas capas, observamos características muy peculiares que nos harán poder resolver este problema de forma bastante rápida y directa gracias a la **regla de la cadena**:
- Al computar la derivada de la función de pérdida en función a los parámetros de la última capa observamos que podemos aplicar fácilmente la regla de la cadena, pues la función de pérdida es, a su vez, función del modelo de la red neuronal, de modo que habrá que multiplicar la derivada de la función de pérdida por la derivada parcial de la propia función en función de los parámetros de esta última capa.
- Para computar las derivadas parciales de la función de pérdida en función a cada parámetro de la penúltima capa, nuevamente tenemos que aplicar la regla de la cadena y observar que sólo necesitamos calcular las derivadas de las funciones de la última capa en función a la penúltima, pues tanto la derivada de la función de pérdida en función a la salida y las derivadas parciales de la salida por cada una de las neuronas de la penúltima capa ya las calculamos en el paso anterior.
- Iterando hacia las demás capas, vemos que no es necesario realizar demasiados cálculos, pues según vamos yendo hacia atrás (**retropropagación**), la mayor parte de los cálculos ya los hemos realizado previamente, por lo que sólo debemos ir añadiendo las derivadas parciales capa a capa.
- Este método de "divide y vencerás" está propiciando que las redes neuronales se comporten de forma muy similar a las neuronas naturales al cumplirse la **regla de Hebb**: aquellas neuronas más fuertemente conectadas (sea signo positivo o negativo) entre sí, experimentarán un mayor cambio cuando las neuronas de la capa posterior queden estimuladas, haciendo que dichas conexiones vayan evolucionando en función a la sincronía de las mismas a la hora de reconocer impulsos.

### 2.2.3.- Funciones de pérdida

La función de pérdidas determina la bondad de ajuste de los parámetros de un modelo y sirve de brújula para decidir con qué parámetros el modelo funciona mejor. Existen varias funciones de pérdidas a considerar, aunque en gran medida depende de si lo que queremos modelizar es una variable numérica o una variable categórica:
- Para variables numéricas suelen utilizarse variantes del **error cuadrático medio** (MSE), escogiendo el tipo de exponente de Minkowski según el grado de robustez que necesitemos.
- Para variables binarias o categóricas suele utilizarse la función **cross-entropy** (CE), que viene a ser el negativo del logaritmo de la verosimilitud de la función softmax como distribución binomial de salida.

### 2.2.4.- Métodos de optimización

La actualización de los pesos del modelo, como hemos visto, es iterativa y se apoya en el descenso por el gradiente.
Esta actualización, sin embargo, no converge siempre de forma eficiente, por lo que a lo largo del tiempo se han ido diseñado diferentes variantes de este método y añadidos para mejorar dicha convergencia.

El primer tipo de mejoras se refiere a la frecuencia de la actualización de los pesos. La opción más evidente sería tomar todos los datos de entrenamiento, medir el error en cada observación, calcular las derivadas por cada peso y realizar la actualización (full gradient). Esta opción no está mal, pero el ritmo de convergencia es lento y necesita mucha computación, por lo que se pasó a probar a hacer una actualización mucho más frecuente: por cada observación, que se escogería al azar para evitar sesgos de ordenación (motivo por el que se llama a esta técnica **Stochastic Gradient Descent, SGD**), de modo que se actualizaría en cada paso. Se trata de un método que funciona mejor que el anterior, pero no tan bien como la tercera opción: llevar a cabo **mini-batches** o lotes aleatorios de unas pocas unidades, decenas o centeneras de observaciones aleatoriamente y trabajando en paralelo, a modo de combinación entre los dos métodos anteriores.

Un segundo tipo de mejoras se centra en incorporar lo que se conce con el término de **momentum**, que viene a ser una especie de "inercia". Tal y como sucede en el método de Newton-Raphson, en el que se introduce la segunda derivada para controlar el ritmo de aprendizaje, al entrenar las redes neuoronales también podemos controlar dicha aceleración incorporando términos que nos acelere o desacelere la tasa de aprendizaje en función cómo ésta va progresando. La técnica de momentum más efectiva es la de Nesterov (NAG), en la que el gradiente se evalúa después de aplicar la velocidad actual a modo de factor de corrección.

Existen otras formas de controlar el ritmo de aprendizaje, entre las que destacan las técnicas **AdaGrad** (adaptive-gradient) en la que la tasa de aprendizaje que aplicamos a para cada parámetro se adapta en función a la variabilidad acumulada que se va observando; **AdaDelta**, que ni siquiera necesita que definamos tasa de aprendizaje porque trata de abordar el problema del descenso paulatino de las tasas de aprendizaje a lo largo del entrenamiento, utilizando ventanas móviles de actualizaciones de gradientes en lugar de acumular todos los gradientes anteriores. Una tercera técnica muy utilizada es **Adam** en la que se adaptan los momentos de primer y segundo orden para obtener a menudo una convergencia más rápida al aprovechar más eficazmente los gradientes más altos de la fase inicial del entrenamiento. Adam parte de la técnica Root Mean Square Propagation (RMSProp), en la que se estima el promedio exponencial de los gradientes en lugar de suma acumulativa de gradientes al cuadrado existente en AdaGrad.

Estas técnicas contribuyen a resolver el principal problema que siempre tuvieron las redes neuronales profundas, la extinción de los errores, que convergen paulatinamente a cero y hacen que la red en cierto modo "se congele". Estas técnicas desde luego contribuyen, pero también han sido fundamentales otras ideas como la selección de las funciones de acivación más adecuadas para cada caso (ReLU, logistic, softmax o Leaky ReLU según el caso) o la **inicialización óptima** de los pesos, siendo la técnica más exitosa la de Xavier Glorot.

### 2.2.5.- Técnicas de regularización

El descubrimiento de las técnicas antes introducidas fueron fundamentales para coseguir entrenar de forma eficiente redes neuronales profundas, pero también jugaron un importante papel las técnicas de regularización para evitar que modelos tan enormes sobreajustaran a las entradas, lo cual habría sido muy decepcionante. Sólo mediante la introducción de técnicas de regularización efectivas, que en cierta forma "dificulten la convergencia inadecuada", se empezarían a conseguir resultados prometedores durante el proceso de entrenamiento.

Aunque ciertas técnias de regularización clásicas de las redes neuronales como la introducción de **ruido aleatorio** o la **parada temprana** llevan utilizándose en las redes neuronales desde antes de los años ochenta para evitar el sobreajuste, lo que marcaría un antes y un después para la convergencia de las redes neuronales profundas sería la introducción de la técnica del **drop-out**. Esta técnica consiste en excluir aleatoriamente y temporalmente algunas neuronas del modelo para evitar que alguna de ellas adquiera demasiada importancia. Esta idea de HInton y su equipo, aunque retarda ligeramente el ritmo de aprendizaje al principio, dota al modelo de una mayor robustez y una mucha mayor solidez para enfrentar al problema del sobreajuste y el de la extinción de los gradientes (vanishing gradients). Cuando aplicamos la técnica del drop-out podemos observar cómo el nivel de acierto en el conjunto de test se mantiene todo el tiempo parejo al del de train, de modo que apenas se produce sobreajuste y, con ello, el modelo se hace mucho más generalizable.

Una técnica parecida, mediante la cual se reduce el número de parámetros y con ello se reduce el sobreajuste, es **replicar pesos** en diferentes partes de la red. Se trata de una técnica que no se debe aplicar teniendo en cuanta la arquitectura de la red y, por ejemplo, es especialmente efectiva en la modelización de las redes siamesas, en las que se suele compartir la totalidad de los pesos en las dos ramas principales de la red.

Por último, otra técnica que ha mostrado ser muy eficaz recientemente es la conocida como **batch normalization**, que consiste en llevar a cabo una normalización periódica de los pesos por capas. Se trata de conseguir un efecto similar al que se busca en otras técnicas de machine learning como el lasso y el ridge, pues así no sólo se combate el sobreajuste sino también otros efectos indeseables como la multicolinealidad de los parámetros que, aunque a efectos de interpretación serían irrelevantes, sí que afecta a nivel de convergencia, especialmente en el conjunto de test.

### 2.2.6.- Descripción del proceso de entrenamiento

Aunque hemos ide describiendo cada uno de los elementos que participan en el entrenamiento por separado, podría ser conveniente hacer una recapitulación y poner en orden los pasos que se han de seguir para llevar a cabo el entrenamiento. Estos serían:
* <ins>Carga</ins> y depuración de datos, detectando la presencia de **missings y outliers**, así como comprobando que disponemos de lo necesario para llevar a cabo el entrenamiento y que la calidad de los datos es la adecuada.
* <ins>Transformación de datos</ins>: aunque en aprendizaje profundo no es indispensable llevar a cabo un proceso de **feature engineer** exhaustivo, suele ser conveniente llevar a cabo pequeñas tareas como generar variables dummy u otro tipo de encodings fuera del modelo.
* <ins>Data augmentation</ins>: a veces es posible aumentar el conjunto de datos mediante diferentes técnicas existentes. Para el caso de la visión por ordenador, tanto TensorFlow como PyTorch disponen de funciones para rotar la imagen, desplazarla, reflectarla, etc. para, de este modo, disponer de una muestra etiquetada más completa.
* <ins>Generación de muestras de datos</ins>: tanto la separación de los datos en conjutos de **train/test** (eventualmente también validación) para supervisar que no existe sobreajuste, hasta otra separación en **mini-batches**, que generalmente se implementa fácilmente mediante las librerías de redes neuronales.
* <ins>Re-utilización de redes neuronales ya existentes cuando sea posible</ins>: esto se conoce como **transfer learning**, de modo que habitualmente no tendremos que construir la red desde cero sino que lo que generalmente se hace es entrenar únicamente la última o últimas capas para adaptarla al dominio de nuestros datos. En internet podemos descargar todo tipo de modelos ya pre-entrenados de forma gratuita y, una vez los tengamos, sólo debemos utilizar el back-bone y añadir sobre ello las capas que queramos.
* <ins>Determinar la arquitectura final de la red</ins>: escogiendo funciones de activación, métodos de regularización y otras opciones, lo cual se puede hacer básicamente por ensayo y error, para lo cual importará nuestra intuición y experiencia, pero que también podrá ser llevado a cabo mediante sistemas automáticos de selección de hiper-parámetros (**AutoML**).
* <ins>Entrenamiento y validación</ins>: cuando ya tengamos todo listo y lancemos el proceso de entrenamiento, iremos viendo cómo progresa el ajuste del modelo tanto en train como en test. Si hemos definido puntos de **check-out**, podremos detener el entrenamiento cuando lo estimemos conveniente y hacer pruebas con él. En cualquier caso, deberemos recordar que será preciso guardar el modelo en disco duro para luego poder utilizarlo.
* <ins>Despliegue y explotación</ins>: una vez lo hayamos validado, podemos testearlo e integrarlo en un proceso productivo, bien sea dentro de un pipeline controlado por terceros o bien desplegándolo por nuestra cuenta para que pueda ser invocado por terceros. En este caso deberemos asegurarnos de que el modelo cumple con las condiciones necesarias para poder ser explotado, incluyendo estabilidad, escalado y que sigue escrupulosamente los estándares de **ética y privacidad** del dato dependiendo del caso.
* <ins>Monitorización y mantenimiento</ins>: una vez desplegado el modelo, debemos asegurarnos de que el modelo no degrada su nivel de acierto por la aparición de **drift** o porque se esté explotando de modo inadecuado. Es conveniente que el modelo esté integrado dentro de un sistema de **DevOps** para que su eventual corrección o mejora sea rápida, sin descuidar tampoco la documentación ni la fase de pruebas.
